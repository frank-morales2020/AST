## mistral-large-latest

```python
import os
from mistralai.client import Mistral

# Retrieve the API key from environment variables
api_key = os.environ["MISTRAL_API_KEY"]

# Initialize the Mistral client
client = Mistral(api_key=api_key)

# Specify the model directly (or fetch via client.models.list())
model = "mistral-large-latest"

# Define the messages for the chat (using supported text format)
messages = [
    {
        "role": "user",
        "content": [
            {
                "type": "text",
                "text": "Hello, can you help me with audio transcription workflows?",
            }
        ],
    }
]

# Get the chat response
chat_response = client.chat.complete(model=model, messages=messages)

# Print the content of the response
print(chat_response.choices[0].message.content)
```

## voxtral-mini-transcribe-3

```python
import os
from mistralai.client import Mistral

# Retrieve the API key from environment variables
api_key = os.environ["MISTRAL_API_KEY"]

# Initialize the Mistral client
client = Mistral(api_key=api_key)

# Open and transcribe the audio file using Mistral's Speech-to-Text API
with open("/content/obama.mp3", "rb") as audio:
    transcription = client.audio.transcriptions.complete(
        model="voxtral-mini-transcribe-3",
        file={"file_name": "obama.mp3", "content": audio},
    )

print("Transcription Result:")
print(transcription.text)
```
