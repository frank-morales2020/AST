```python
def github_repo_list(repo):
  import requests

  repo = "frank-morales2020/AST"
  branch = "main"

  url = f"https://api.github.com/repos/{repo}/git/trees/{branch}?recursive=1"
  response = requests.get(url).json()

  notebooks = [
      item["path"]
      for item in response.get("tree", [])
      if item["path"].endswith(".ipynb")
  ]

  for nb in notebooks:
      print(nb)
```

```python
def github_repo_analysis(repo):
  from collections import Counter
  import requests


  #repo = "frank-morales2020/MLxDL"
  branch = "main"

  url = f"https://api.github.com/repos/{repo}/git/trees/{branch}?recursive=1"
  response = requests.get(url).json()

  # Extract all .ipynb paths
  notebooks = [
      item["path"]
      for item in response.get("tree", [])
      if item["path"].endswith(".ipynb")
  ]

  # 1. Directory / Folder Distribution
  directories = Counter(
      nb.split("/")[0] if "/" in nb else "Root" for nb in notebooks
  )

  # 2. Keyword Frequency Analysis
  keywords = ["TOPO", "GEMMA", "KIMI", "LEFM", "AGENTIC", "H2E", "DEEPSEEK", "MISTRAL"]
  keyword_counts = {
      kw: sum(1 for nb in notebooks if kw in nb.upper()) for kw in keywords
  }

  print("=== REPOSITORY ANALYTICS SUMMARY ===")
  print(f"Total Notebooks Found: {len(notebooks)}\n")

  print("--- Folder Breakdown ---")
  for folder, count in directories.most_common():
      print(f"  {folder}: {count} notebooks")

  print("\n--- Keyword Mentions ---")
  for kw, count in sorted(keyword_counts.items(), key=lambda x: x[1], reverse=True):
      print(f"  {kw}: {count} files")
```

```python
github_repo_list("frank-morales2020/MLxDL")
```

```python
github_repo_analysis("frank-morales2020/MLxDL")
```

```python
github_repo_list("frank-morales2020/AST")
```

```python
github_repo_analysis("frank-morales2020/AST")
```

## git-clone

```python
import os
import subprocess
import shutil

repo_url = "https://github.com/frank-morales2020/AST.git"
repo_name = "AST"

# Clone repository if it doesn't exist yet
if not os.path.exists(repo_name):
    print(f"Cloning {repo_url}...")
    subprocess.run(["git", "clone", repo_url], check=True)
else:
    print(f"Directory '{repo_name}' already exists. Skipping clone.")

# Move into the repository directory
os.chdir(repo_name)
print(f"Current Working Directory: {os.getcwd()}")

# Define categories
categories = {
    "TOPO": "TOPO_Notebooks",
    "GEMMA": "Gemma_Notebooks",
    "H2E": "H2E_Notebooks",
    "LEFM": "LEFM_Notebooks",
    "KIMI": "Kimi_Notebooks",
    "AGENTIC": "Agentic_Notebooks"
}

# Create destination folders
for folder in categories.values():
    os.makedirs(folder, exist_ok=True)

# Scan and sort notebooks
files = os.listdir(".")
notebooks = [f for f in files if f.endswith(".ipynb") and os.path.isfile(f)]

print(f"Found {len(notebooks)} notebook(s) to organize.\n")

for filename in notebooks:
    upper_name = filename.upper()
    moved = False

    for keyword, folder in categories.items():
        if keyword in upper_name:
            dest_path = os.path.join(folder, filename)
            shutil.move(filename, dest_path)
            print(f"Moved: {filename} -> {folder}/")
            moved = True
            break

    if not moved:
        print(f"Left in root: {filename}")

print("\nOrganization complete! You can now commit and push these changes from the 'AST' directory.")
```
