## 📊 FIGURE 1: The Topological Governor Mechanism

```python
%matplotlib inline
import warnings
warnings.filterwarnings('ignore', category=UserWarning, module='matplotlib.font_manager')

import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np
from matplotlib.patches import Patch

# Define primes
PRIME_ANCHORS = [2, 3, 5, 7, 11, 13]
GRID_SIZE = 14

# Colors
RED = '#E74C3C'
DARK_RED = '#C0392B'
BLUE = '#3498DB'
GREEN = '#27AE60'
LIGHT_BLUE = '#5DADE2'
GOLD = '#F1C40F'
GRAY = '#BDC3C7'
DARK_GRAY = '#7F8C8D'
DARK_BLUE = '#2C3E50'
WHITE = '#FFFFFF'

# Create figure
fig, axes = plt.subplots(1, 3, figsize=(18, 5.5))
fig.suptitle('Figure 1: The Topological Governor Mechanism', fontsize=16, fontweight='bold')

# PANEL A
ax1 = axes[0]
ax1.set_title('A: Before Training (Snapshot)', fontsize=11, fontweight='bold')
ax1.set_xlim(0, GRID_SIZE)
ax1.set_ylim(GRID_SIZE, 0)

for i in range(GRID_SIZE):
    for j in range(GRID_SIZE):
        if i in PRIME_ANCHORS and j in PRIME_ANCHORS:
            ax1.add_patch(patches.Rectangle((i, j), 1.0, 1.0, facecolor=RED, alpha=0.85, edgecolor=DARK_RED, linewidth=2))
        else:
            ax1.add_patch(patches.Rectangle((i, j), 1.0, 1.0, facecolor=BLUE, alpha=0.25, edgecolor=GRAY, linewidth=0.5))

for i in PRIME_ANCHORS:
    for j in PRIME_ANCHORS:
        ax1.text(i + 0.5, j + 0.5, '★', ha='center', va='center', fontsize=12, color=WHITE, fontweight='bold')

ax1.legend([Patch(facecolor=RED, alpha=0.85), Patch(facecolor=BLUE, alpha=0.25)],
           ['Prime Anchors', 'Adaptable'], loc='lower center', fontsize=8, ncol=2, frameon=True, bbox_to_anchor=(0.5, -0.15))
ax1.set_xlabel('Embedding Dimension', fontsize=10)
ax1.set_ylabel('Embedding Dimension', fontsize=10)
ax1.set_xticks([])
ax1.set_yticks([])
ax1.set_aspect('equal')
ax1.spines['top'].set_visible(True)
ax1.spines['right'].set_visible(True)
ax1.spines['bottom'].set_visible(True)
ax1.spines['left'].set_visible(True)

# PANEL B
ax2 = axes[1]
ax2.set_title('B: During Training (Gradient Zeroing)', fontsize=11, fontweight='bold')
ax2.set_xlim(0, GRID_SIZE)
ax2.set_ylim(GRID_SIZE, 0)

for i in range(GRID_SIZE):
    for j in range(GRID_SIZE):
        if i in PRIME_ANCHORS and j in PRIME_ANCHORS:
            ax2.add_patch(patches.Rectangle((i, j), 1.0, 1.0, facecolor=RED, alpha=0.85, edgecolor=DARK_RED, linewidth=2))
        else:
            ax2.add_patch(patches.Rectangle((i, j), 1.0, 1.0, facecolor=GREEN, alpha=0.35, edgecolor=DARK_GRAY, linewidth=0.5))

for i in PRIME_ANCHORS:
    for j in PRIME_ANCHORS:
        ax2.text(i + 0.5, j + 0.5, '✕', ha='center', va='center', fontsize=16, color=WHITE, fontweight='bold')

np.random.seed(42)
for i in range(GRID_SIZE):
    for j in range(GRID_SIZE):
        if not (i in PRIME_ANCHORS and j in PRIME_ANCHORS):
            dx, dy = 0.25 * np.random.randn(), 0.25 * np.random.randn()
            magnitude = np.hypot(dx, dy)
            if magnitude > 0.05:
                ax2.arrow(i + 0.5, j + 0.5, dx, dy, head_width=0.1, head_length=0.1, fc=GREEN, ec=GREEN, alpha=0.6)

ax2.legend([Patch(facecolor=RED, alpha=0.85), Patch(facecolor=GREEN, alpha=0.35)],
           ['Frozen Anchors', 'Adapting'], loc='lower center', fontsize=8, ncol=2, frameon=True, bbox_to_anchor=(0.5, -0.15))
ax2.set_xlabel('Embedding Dimension', fontsize=10)
ax2.set_xticks([])
ax2.set_yticks([])
ax2.set_aspect('equal')
ax2.spines['top'].set_visible(True)
ax2.spines['right'].set_visible(True)
ax2.spines['bottom'].set_visible(True)
ax2.spines['left'].set_visible(True)

# PANEL C
ax3 = axes[2]
ax3.set_title('C: After Training (Enforce Anchors)', fontsize=11, fontweight='bold')
ax3.set_xlim(0, GRID_SIZE)
ax3.set_ylim(GRID_SIZE, 0)

for i in range(GRID_SIZE):
    for j in range(GRID_SIZE):
        if i in PRIME_ANCHORS and j in PRIME_ANCHORS:
            ax3.add_patch(patches.Rectangle((i, j), 1.0, 1.0, facecolor=RED, alpha=0.9, edgecolor=GOLD, linewidth=2.5))
        else:
            ax3.add_patch(patches.Rectangle((i, j), 1.0, 1.0, facecolor=LIGHT_BLUE, alpha=0.5, edgecolor=GRAY, linewidth=0.5))

for i in PRIME_ANCHORS:
    for j in PRIME_ANCHORS:
        ax3.text(i + 0.5, j + 0.5, '✓', ha='center', va='center', fontsize=14, color=WHITE, fontweight='bold')

np.random.seed(42)
for i in range(GRID_SIZE):
    for j in range(GRID_SIZE):
        if not (i in PRIME_ANCHORS and j in PRIME_ANCHORS):
            if np.random.random() > 0.6:
                ax3.text(i + 0.5, j + 0.5, '•', ha='center', va='center', fontsize=10, color=DARK_BLUE, alpha=0.5)

ax3.legend([Patch(facecolor=RED, alpha=0.9, edgecolor=GOLD), Patch(facecolor=LIGHT_BLUE, alpha=0.5)],
           ['Anchors Restored', 'New Knowledge'], loc='lower center', fontsize=8, ncol=2, frameon=True, bbox_to_anchor=(0.5, -0.15))
ax3.set_xlabel('Embedding Dimension', fontsize=10)
ax3.set_xticks([])
ax3.set_yticks([])
ax3.set_aspect('equal')
ax3.spines['top'].set_visible(True)
ax3.spines['right'].set_visible(True)
ax3.spines['bottom'].set_visible(True)
ax3.spines['left'].set_visible(True)

# Clean layout
plt.subplots_adjust(left=0.08, right=0.95, top=0.88, bottom=0.32, wspace=0.35)
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

fig5, axes = plt.subplots(2, 2, figsize=(15, 10))
fig5.suptitle('Figure 5: TOPO-2026 - Universal Architecture-Agnostic Validation',
              fontsize=16, fontweight='bold')

# Data from your leaderboard
architectures = ['Vision Transformer', 'StripedHyena', 'GLM Transformer',
                'Sparse MoE FP8', 'Fine-grained MoE FP8', 'Hybrid Attention',
                'MoE Architecture', 'Dense BF16 Transformer', 'Sparse MoE FP8',
                'Dense Transformer + JEPA']
models = ['Gemma-4', 'Evo 2', 'GLM-4.6V', 'Sarvam-30B', 'DeepSeek',
          'Muse', 'Kimi 2', 'GPT-OSS', 'Mixtral', 'TOPO-JEPA']
accuracy = [100.00, 99.73, 97.50, 95.90, 95.40, 94.20, 93.85, 92.30, 89.70, 89.00]
fgt = [0.00, 0.83, 2.10, -0.60, 0.03, 0.45, -0.12, 1.55, -1.85, -0.75]

# Architecture categories
arch_categories = ['Transformer', 'Hyena', 'Transformer', 'MoE', 'MoE',
                   'Hybrid', 'MoE', 'Transformer', 'MoE', 'Transformer']
colors_arch = {'Transformer': '#1f77b4', 'Hyena': '#ff7f0e', 'MoE': '#2ca02c', 'Hybrid': '#d62728'}

# Panel A: Accuracy by Architecture
ax1 = axes[0, 0]
ax1.set_title('A: Task C Accuracy by Architecture', fontsize=14)

# Group by architecture
unique_archs = list(set(arch_categories))
positions = np.arange(len(unique_archs))
means = []
stds = []

for arch in unique_archs:
    idx = [i for i, a in enumerate(arch_categories) if a == arch]
    acc_values = [accuracy[i] for i in idx]
    means.append(np.mean(acc_values))
    stds.append(np.std(acc_values))

bars = ax1.bar(unique_archs, means, yerr=stds, capsize=5, color=[colors_arch[a] for a in unique_archs])
ax1.set_ylabel('Task C Accuracy (%)', fontsize=12)
ax1.set_ylim(85, 102)
ax1.axhline(y=95.77, color='blue', linestyle=':', linewidth=2, label='Average: 95.77%')
ax1.axhline(y=85, color='red', linestyle='--', linewidth=2, label='Threshold: 85%')
ax1.legend()
ax1.grid(True, alpha=0.3)

# Add values on bars
for i, (bar, mean) in enumerate(zip(bars, means)):
    ax1.text(i, mean + 0.5, f'{mean:.1f}%', ha='center', fontsize=10, fontweight='bold')

# Panel B: Forgetting by Architecture
ax2 = axes[0, 1]
ax2.set_title('B: Forgetting by Architecture', fontsize=14)

means_fgt = []
stds_fgt = []
for arch in unique_archs:
    idx = [i for i, a in enumerate(arch_categories) if a == arch]
    fgt_values = [fgt[i] for i in idx]
    means_fgt.append(np.mean(fgt_values))
    stds_fgt.append(np.std(fgt_values))

bars2 = ax2.bar(unique_archs, means_fgt, yerr=stds_fgt, capsize=5, color=[colors_arch[a] for a in unique_archs])
ax2.set_ylabel('Forgetting (FGT %)', fontsize=12)
ax2.axhline(y=0.49, color='blue', linestyle=':', linewidth=2, label='Average: 0.49%')
ax2.axhline(y=10, color='red', linestyle='--', linewidth=2, label='Threshold: 10%')
ax2.axhline(y=0, color='black', linestyle='-', linewidth=1)
ax2.legend()
ax2.grid(True, alpha=0.3)

# Add values on bars
for i, (bar, mean) in enumerate(zip(bars2, means_fgt)):
    ax2.text(i, mean + 0.2 if mean >= 0 else mean - 0.5,
             f'{mean:.2f}%', ha='center', fontsize=10, fontweight='bold')

# Panel C: Certification Rate by Architecture
ax3 = axes[1, 0]
ax3.set_title('C: 100% Certification Rate Across All Architectures', fontsize=14)

# Data
arch_names = ['Striped\nHyena', 'Vision\nTransformer', 'GLM\nTransformer',
              'Sparse\nMoE', 'Fine-grained\nMoE', 'Hybrid\nAttention',
              'MoE', 'Dense\nTransformer', 'Dense\n+ JEPA']
success_rates = [100] * len(arch_names)

bars3 = ax3.barh(arch_names, success_rates, color='green', alpha=0.7)
ax3.set_xlabel('Certification Rate (%)', fontsize=12)
ax3.set_xlim(0, 105)
ax3.axvline(x=100, color='red', linestyle='--', linewidth=2, label='100% Certification')
ax3.legend()
ax3.grid(True, alpha=0.3)

# Add labels
for i, (bar, rate) in enumerate(zip(bars3, success_rates)):
    ax3.text(rate + 0.5, i, f'{rate}%', va='center', fontsize=10, fontweight='bold')

# Panel D: Summary Metrics
ax4 = axes[1, 1]
ax4.set_title('D: Summary Statistics', fontsize=14)
ax4.axis('off')

# Create a table
metrics = [
    ['Metric', 'Value'],
    ['Total Models', '10'],
    ['Architectures', '7'],
    ['Certification Rate', '100%'],
    ['Avg Task C Accuracy', '95.77%'],
    ['Avg Forgetting', '0.49%'],
    ['Best Task C', '100.00% (Gemma-4)'],
    ['Best FGT', '0.00% (Gemma-4)'],
    ['First Non-Transformer', 'Evo 2 (StripedHyena)'],
    ['Evo 2 Accuracy', '99.73%'],
    ['Evo 2 FGT', '0.83%']
]

# Create table with colors
table = ax4.table(cellText=metrics, loc='center', cellLoc='center',
                  colWidths=[0.3, 0.6])
table.auto_set_font_size(False)
table.set_fontsize(11)
table.scale(1.2, 1.5)

# Color header
for i in range(2):
    table[(0, i)].set_facecolor('#1f77b4')
    table[(0, i)].set_text_props(weight='bold', color='white')

# Highlight Evo2
for i in range(2):
    table[(9, i)].set_facecolor('#ff7f0e')
    table[(9, i)].set_text_props(weight='bold')

# Highlight Gemma-4
for i in range(2):
    table[(7, i)].set_facecolor('#2ca02c')
    table[(7, i)].set_text_props(weight='bold')

plt.tight_layout()
plt.savefig('Figure5_Universal_Validation.png', dpi=300, bbox_inches='tight')
plt.show()
```
