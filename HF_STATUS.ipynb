````python
from google.colab import userdata
from huggingface_hub import HfApi
import pandas as pd
from datetime import datetime, timezone
import requests
import re

# Retrieve token and initialize API
HF_TOKEN = userdata.get('HF_TOKEN')
username = 'frankmorales2020'
api = HfApi(token=HF_TOKEN)

# Fetch all models with full metadata
models = api.list_models(author=username, full=True)

def analyze_readme_quality(content):
    """Analyze README quality with robust detection for all documentation styles"""
    quality = {
        'has_model_description': False,
        'has_technical_details': False,
        'has_training_code': False,
        'has_dataset_info': False,
        'has_eval_results': False,
        'has_inference_code': False,
        'has_expected_output': False,
        'has_github_link': False,
        'has_paper_link': False,
        'has_certificate': False,
        'has_author_info': False,
        'content_length': len(content),
        'description_quality': 'none'  # 'none', 'basic', 'good', 'excellent'
    }

    if not content:
        return quality

    content_lower = content.lower()

    # ===== MODEL DESCRIPTION =====
    # Check for title or description patterns
    desc_patterns = [
        r'^#.*?(?:model|topological|gpt|llama|mistral|bert|gemma|deepseek|mixtral|kimi|evo2)',  # Title with model name
        r'#\s*(?:model|description|overview)',
        r'##\s*(?:model|description|overview)',
        r'author:',
        r'paper:',
        r'certificate:',
        r'finetuned on',
        r'trained on',
        r'fine-tuned on',
        r'^.*?(?:is a|are a|this model).*?(?:for|that)',
        r'#.*?Description.*?\n.*?[A-Z]',
        r'```.*?\n.*?description',
        r'Model Description:',
        r'What is this model',
        r'# .*?for .*?tasks',
        r'# .*?—.*?'  # Title with em dash
    ]
    for pattern in desc_patterns:
        if re.search(pattern, content, re.IGNORECASE):
            quality['has_model_description'] = True
            break

    # If content has any substantial text beyond code blocks, consider it a description
    text_lines = [line for line in content.split('\n') if line.strip() and not line.strip().startswith('```')]
    text_content = ' '.join(text_lines)

    if len(text_content) > 100:  # Arbitrary threshold for meaningful text
        quality['has_model_description'] = True
        if len(text_content) > 300:
            quality['description_quality'] = 'excellent'
        elif len(text_content) > 150:
            quality['description_quality'] = 'good'
        else:
            quality['description_quality'] = 'basic'

    # ===== AUTHOR INFO =====
    author_patterns = [
        r'author:',
        r'authors?:',
        r'by\s+[A-Z][a-z]+ [A-Z][a-z]+',
        r'lab:',
        r'institution:',
        r'university:',
        r'@.*\.com'
    ]
    for pattern in author_patterns:
        if re.search(pattern, content, re.IGNORECASE):
            quality['has_author_info'] = True
            break

    # ===== CERTIFICATE =====
    cert_patterns = [
        r'certificate',
        r'certified',
        r'track ii',
        r'topo-2026',
        r'pass',
        r'threshold:',
        r'combined model forgetting',
        r'topological anchor memory',
        r'empirical execution ledger'
    ]
    for pattern in cert_patterns:
        if re.search(pattern, content, re.IGNORECASE):
            quality['has_certificate'] = True
            break

    # ===== TECHNICAL DETAILS =====
    tech_patterns = [
        r'version',
        r'requires?:',
        r'install',
        r'torch',
        r'transformers',
        r'unsloth',
        r'bitsandbytes',
        r'cuda',
        r'quantization',
        r'boundary layer',
        r'prime anchor',
        r'kv cache',
        r'flash attention',
        r'peft',
        r'load_in_4bit',
        r'load_in_8bit',
        r'bf16',
        r'fp16',
        r'dtype',
        r'nvidia',
        r'gpu',
        r'safetensors',
        r'checkpoint',
        r'config\.json',
        r'batch size',
        r'learning rate',
        r'optimizer',
        r'model.*parameters',
        r'parameters:',
        r'layers?:',
        r'hidden size',
        r'attention heads',
        r'context length'
    ]
    for pattern in tech_patterns:
        if re.search(pattern, content, re.IGNORECASE):
            quality['has_technical_details'] = True
            break

    # If there's a certificate, consider it as technical details
    if quality['has_certificate']:
        quality['has_technical_details'] = True

    # ===== TRAINING CODE =====
    training_patterns = [
        r'github\.com/.*\.py',
        r'github\.com/.*\.ipynb',
        r'```python.*?train',
        r'from.*import.*train',
        r'def train',
        r'class.*Trainer',
        r'fine-?tun',
        r'training',
        r'epoch',
        r'learning rate',
        r'batch size',
        r'training loop',
        r'train\.py',
        r'fine_tune\.py',
        r'trainer',
        r'```bash.*?python.*train',
        r'```.*?training',
        r'sft_trainer',
        r'loss',
        r'optimizer'
    ]
    for pattern in training_patterns:
        if re.search(pattern, content, re.IGNORECASE):
            quality['has_training_code'] = True
            break

    # ===== DATASET INFO =====
    dataset_patterns = [
        r'dataset',
        r'data:',
        r'training data',
        r'corpus',
        r'benchmark',
        r'used.*dataset',
        r'datasets?\.load',
        r'from datasets import',
        r'train.*data',
        r'validation.*data',
        r'test.*data',
        r'squad',
        r'glue',
        r'superglue',
        r'imagenet',
        r'mnist',
        r'cifar',
        r'wikipedia',
        r'bookcorpus',
        r'mc4',
        r'thesaurus',
        r'wordnet',
        r'onto notes',
        r'conll',
        r'task.*a.*world',
        r'task.*b.*business',
        r'task.*c.*sci'
    ]
    for pattern in dataset_patterns:
        if re.search(pattern, content, re.IGNORECASE):
            quality['has_dataset_info'] = True
            break

    # ===== EVALUATION RESULTS =====
    eval_patterns = [
        r'accuracy',
        r'f1',
        r'bleu',
        r'rouge',
        r'perplexity',
        r'eval',
        r'benchmark',
        r'score',
        r'result',
        r'%',
        r'✅.*%',
        r'0% forgetting',
        r'pass',
        r'empirical execution',
        r'task.*performance',
        r'metrics?',
        r'leaderboard',
        r'precision',
        r'recall',
        r'auc',
        r'roc',
        r'loss',
        r'pass@',
        r'mmlu',
        r'hellaswag',
        r'gsm8k',
        r'math',
        r'humanval',
        r'mbpp',
        r'codexglue',
        r'retention drift',
        r'programmatic factual accuracy'
    ]
    for pattern in eval_patterns:
        if re.search(pattern, content, re.IGNORECASE):
            quality['has_eval_results'] = True
            break

    # If there's a certificate with metrics, consider it as evaluation results
    if quality['has_certificate'] and ('accuracy' in content_lower or '%' in content):
        quality['has_eval_results'] = True

    # ===== INFERENCE CODE =====
    inference_patterns = [
        r'```python',
        r'from.*import.*model',
        r'generate',
        r'predict',
        r'inference',
        r'from_pretrained',
        r'hf_hub_download',
        r'model\.generate',
        r'tokenizer\.decode',
        r'pipeline',
        r'inference code',
        r'usage',
        r'example.*usage',
        r'how to use',
        r'quick start',
        r'load.*model',
        r'usage example'
    ]
    for pattern in inference_patterns:
        if re.search(pattern, content, re.IGNORECASE):
            quality['has_inference_code'] = True
            break

    # ===== EXPECTED OUTPUT =====
    output_patterns = [
        r'expected.*output',
        r'output:',
        r'```.*output',
        r'\[task.*\]',
        r'results:',
        r'should see',
        r'should output',
        r'will output',
        r'outputs:',
        r'example output',
        r'sample output',
        r'generates:',
        r'➔'
    ]
    for pattern in output_patterns:
        if re.search(pattern, content, re.IGNORECASE):
            quality['has_expected_output'] = True
            break

    # ===== GITHUB LINK =====
    github_patterns = [
        r'github\.com',
        r'gitlab\.com'
    ]
    for pattern in github_patterns:
        if re.search(pattern, content, re.IGNORECASE):
            quality['has_github_link'] = True
            break

    # ===== PAPER LINK =====
    paper_patterns = [
        r'zenodo\.org',
        r'arxiv\.org',
        r'doi\.org',
        r'paper:',
        r'publication:',
        r'\d{4}\.\d{5}',
        r'https?://.*\.pdf',
        r'research paper',
        r'preprint'
    ]
    for pattern in paper_patterns:
        if re.search(pattern, content, re.IGNORECASE):
            quality['has_paper_link'] = True
            break

    return quality

# Build dataset with accurate analysis
model_analysis = []

for m in models:
    model_id = m.modelId
    downloads = getattr(m, 'downloads', 0)
    likes = getattr(m, 'likes', 0)
    is_private = getattr(m, 'private', False)
    tags = getattr(m, 'tags', [])
    library_name = getattr(m, 'library_name', None)
    last_modified = getattr(m, 'lastModified', None)

    # Check age
    age_days = None
    if last_modified:
        age_days = (datetime.now(timezone.utc) - last_modified).days

    # Get README content
    readme_content = ""
    has_readme = False
    try:
        url = f"https://huggingface.co/{model_id}/raw/main/README.md"
        response = requests.get(url, headers={"Authorization": f"Bearer {HF_TOKEN}"}, timeout=10)
        if response.status_code == 200:
            readme_content = response.text
            has_readme = True
    except:
        pass

    # Analyze README quality
    readme_quality = analyze_readme_quality(readme_content)

    # Calculate seriousness score with improved weighting
    seriousness_score = 0

    # Core requirements (high weight)
    if readme_quality['has_model_description']:
        seriousness_score += 2
    if readme_quality['has_technical_details']:
        seriousness_score += 2
    if readme_quality['has_training_code']:
        seriousness_score += 2
    if readme_quality['has_eval_results']:
        seriousness_score += 2

    # Nice to have (medium weight)
    if readme_quality['has_inference_code']:
        seriousness_score += 1
    if readme_quality['has_expected_output']:
        seriousness_score += 1
    if readme_quality['has_dataset_info']:
        seriousness_score += 1
    if readme_quality['has_github_link']:
        seriousness_score += 1
    if readme_quality['has_paper_link']:
        seriousness_score += 1
    if readme_quality['has_author_info']:
        seriousness_score += 1
    if readme_quality['has_certificate']:
        seriousness_score += 1

    # Community metrics (low weight)
    if downloads > 10:
        seriousness_score += 1
    if likes > 0:
        seriousness_score += 1
    if not is_private:
        seriousness_score += 1
    if library_name:
        seriousness_score += 1

    # Determine seriousness level with adjusted thresholds
    if seriousness_score >= 10:
        seriousness_level = "🏆 Gold Standard (Excellent)"
    elif seriousness_score >= 7:
        seriousness_level = "✅ Serious Researcher/Practitioner"
    elif seriousness_score >= 4:
        seriousness_level = "⚠️ Amateur or Experimenter"
    else:
        seriousness_level = "❌ Not Serious (or just backing up)"

    # Compile red flags
    red_flags = []
    if not readme_quality['has_model_description']:
        red_flags.append("No model description")
    if not readme_quality['has_technical_details']:
        red_flags.append("No technical details")
    if not readme_quality['has_training_code']:
        red_flags.append("No training code/params")
    if not readme_quality['has_dataset_info']:
        red_flags.append("No dataset info")
    if not readme_quality['has_eval_results']:
        red_flags.append("No evaluation results")
    if not readme_quality['has_inference_code']:
        red_flags.append("No inference code")
    if not readme_quality['has_expected_output']:
        red_flags.append("No expected output")
    if not readme_quality['has_github_link']:
        red_flags.append("No GitHub link")
    if not readme_quality['has_paper_link']:
        red_flags.append("No paper link")
    if not readme_quality['has_author_info']:
        red_flags.append("No author info")
    if is_private:
        red_flags.append("Private repository")
    if age_days and age_days > 180 and downloads < 5:
        red_flags.append("Old with no downloads")

    model_analysis.append({
        'model_id': model_id,
        'private': is_private,
        'downloads': downloads,
        'likes': likes,
        'age_days': age_days,
        'readme_length': len(readme_content),
        'has_readme': has_readme,
        'has_model_description': readme_quality['has_model_description'],
        'has_author_info': readme_quality['has_author_info'],
        'has_certificate': readme_quality['has_certificate'],
        'has_technical_details': readme_quality['has_technical_details'],
        'has_training_code': readme_quality['has_training_code'],
        'has_dataset_info': readme_quality['has_dataset_info'],
        'has_eval_results': readme_quality['has_eval_results'],
        'has_inference_code': readme_quality['has_inference_code'],
        'has_expected_output': readme_quality['has_expected_output'],
        'has_github_link': readme_quality['has_github_link'],
        'has_paper_link': readme_quality['has_paper_link'],
        'description_quality': readme_quality['description_quality'],
        'seriousness_score': seriousness_score,
        'seriousness_level': seriousness_level,
        'red_flags': ', '.join(red_flags) if red_flags else 'None',
        'num_red_flags': len(red_flags),
        'library_name': library_name,
        'last_modified': last_modified,
        'tags': ', '.join(tags[:5]) + ('...' if len(tags) > 5 else '')
    })

# Create DataFrame
df_analysis = pd.DataFrame(model_analysis)

# Display summary statistics
print("="*80)
print("COMPLETE SERIOUSNESS ANALYSIS OF MODELS")
print("="*80)
print(f"\nTotal models analyzed: {len(df_analysis)}")
print("\nBreakdown by seriousness level:")
print(df_analysis['seriousness_level'].value_counts().to_string())

print("\n" + "="*80)
print("TOP 15 MOST SERIOUS MODELS (by score)")
print("="*80)
top_serious = df_analysis.sort_values('seriousness_score', ascending=False).head(15)
display(top_serious[['model_id', 'seriousness_score', 'seriousness_level', 'downloads', 'likes', 'num_red_flags']])

# Find specific models
specific_models = [
    'frankmorales2020/topo-gemma-4-e4b-vision-13tasks',
    'frankmorales2020/topological-ai-gpt-oss-20b'
]

for specific_model_id in specific_models:
    specific_model = df_analysis[df_analysis['model_id'] == specific_model_id]
    if not specific_model.empty:
        print("\n" + "="*80)
        print(f"ANALYSIS OF: {specific_model_id}")
        print("="*80)
        display(specific_model[[
            'model_id', 'seriousness_score', 'seriousness_level',
            'red_flags', 'num_red_flags', 'has_model_description',
            'has_author_info', 'has_certificate', 'has_technical_details',
            'has_training_code', 'has_dataset_info', 'has_eval_results',
            'has_inference_code', 'has_expected_output', 'has_github_link',
            'has_paper_link', 'description_quality'
        ]])

print("\n" + "="*80)
print("RED FLAG STATISTICS")
print("="*80)
red_flag_counts = {
    'No model description': len(df_analysis[~df_analysis['has_model_description']]),
    'No author info': len(df_analysis[~df_analysis['has_author_info']]),
    'No certificate': len(df_analysis[~df_analysis['has_certificate']]),
    'No technical details': len(df_analysis[~df_analysis['has_technical_details']]),
    'No training code': len(df_analysis[~df_analysis['has_training_code']]),
    'No dataset info': len(df_analysis[~df_analysis['has_dataset_info']]),
    'No evaluation results': len(df_analysis[~df_analysis['has_eval_results']]),
    'No inference code': len(df_analysis[~df_analysis['has_inference_code']]),
    'No expected output': len(df_analysis[~df_analysis['has_expected_output']]),
    'No GitHub link': len(df_analysis[~df_analysis['has_github_link']]),
    'No paper link': len(df_analysis[~df_analysis['has_paper_link']]),
}
for flag, count in sorted(red_flag_counts.items(), key=lambda x: x[1], reverse=True):
    percentage = (count / len(df_analysis)) * 100
    print(f"  {flag}: {count} models ({percentage:.1f}%)")

# Show models that need improvement
print("\n" + "="*80)
print("MODELS NEEDING IMPROVEMENT (Score < 7)")
print("="*80)
needs_improvement = df_analysis[df_analysis['seriousness_score'] < 7]
if not needs_improvement.empty:
    display(needs_improvement[[
        'model_id', 'seriousness_score', 'seriousness_level',
        'num_red_flags', 'has_model_description', 'has_technical_details',
        'has_training_code', 'has_eval_results', 'has_inference_code'
    ]])
else:
    print("  No models need improvement! All are at least 'Serious' quality.")

# Show best and worst examples
print("\n" + "="*80)
print("BEST PRACTICE EXAMPLE (Highest Quality)")
print("="*80)
best_model = df_analysis[df_analysis['seriousness_score'] >= 14].head(1)
if not best_model.empty:
    display(best_model[[
        'model_id', 'seriousness_score', 'seriousness_level',
        'red_flags', 'has_model_description', 'has_author_info',
        'has_technical_details', 'has_training_code', 'has_eval_results',
        'has_inference_code', 'has_github_link', 'has_paper_link'
    ]])

print("\n" + "="*80)
print("WORST PRACTICE EXAMPLE (Lowest Quality)")
print("="*80)
worst_model = df_analysis[df_analysis['seriousness_score'] < 4].head(1)
if not worst_model.empty:
    display(worst_model[[
        'model_id', 'seriousness_score', 'seriousness_level',
        'red_flags', 'has_model_description', 'has_technical_details',
        'has_training_code', 'has_eval_results', 'has_inference_code',
        'has_github_link', 'has_paper_link'
    ]])

# Export to CSV
df_analysis.to_csv('frankmorales2020_models_seriousness_analysis.csv', index=False)
print("\n" + "="*80)
print(f"Full analysis exported to: frankmorales2020_models_seriousness_analysis.csv")
print("="*80)
````
