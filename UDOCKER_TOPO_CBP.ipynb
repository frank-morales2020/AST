## UDOCKER

```python
%%bash
# 1. Clean up stale containers
udocker --allow-root rm ollama-runner 2>/dev/null || true

# 2. Python script using standard os.environ for token retrieval
python3 -c '
import os
from huggingface_hub import HfApi, hf_hub_download, login

hf_token = os.environ.get("HF_TOKEN")
if hf_token:
    login(token=hf_token)
    print("[+] Authenticated with Hugging Face via environment.")

repo_id = "frankmorales2020/certified-scorpion-topo-cbp-qwen-master-GGUF"
api = HfApi()
files = api.list_repo_files(repo_id=repo_id, repo_type="model")

gguf_files = [f for f in files if f.endswith(".gguf")]
filename = gguf_files[0]

os.makedirs("./weights", exist_ok=True)
model_path = hf_hub_download(repo_id=repo_id, filename=filename, local_dir="./weights")

modelfile_content = f"""FROM ./weights/{filename}
PARAMETER temperature 0.0
PARAMETER seed 123
PARAMETER num_ctx 1024
PARAMETER num_thread 2
SYSTEM "You are a deterministic cognitive engineering assistant governed by Topological Continual Backpropagation (Topo-CBP)."
"""

with open("./Modelfile.scorpion", "w") as f:
    f.write(modelfile_content)
print(f"[+] Prepared Modelfile for {filename}.")
'

# 3. Pull base image first, then create and configure container
udocker --allow-root pull ubuntu:22.04
udocker --allow-root create --name=ollama-runner ubuntu:22.04
udocker --allow-root setup --execmode=P1 ollama-runner

# 4. Execute inside udocker container
udocker --allow-root run --volume=$(pwd):/workspace --workdir=/workspace ollama-runner /bin/bash -c "
export DEBIAN_FRONTEND=noninteractive
apt-get update && apt-get install -y curl ca-certificates zstd >/dev/null 2>&1
curl -fsSL https://ollama.com/install.sh | sh >/dev/null 2>&1

ollama serve > /tmp/ollama.log 2>&1 &
until curl -s http://127.0.0.1:11434/ >/dev/null; do sleep 1; done

echo '[+] Ollama server active inside udocker. Building model...'
ollama create certified-scorpion -f /workspace/Modelfile.scorpion

echo '[+] Model built successfully. Running verification query...'
ollama run certified-scorpion 'Verify topological invariant stability under seed 123.'
"
```
