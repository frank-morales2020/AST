```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Complex Plane Setup (Unit Circle)
theta = np.linspace(0, 2 * np.pi, 500)
z = np.exp(1j * theta)

# Create figure with two subplots
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))

# Plot 1: Unit circle and phasor representation
ax1.plot(z.real, z.imag, label='e^(iθ) Unit Circle', color='royalblue', lw=2)

# Highlight a specific angle, e.g., theta = pi/4 (45 degrees)
th_sample = np.pi / 4
z_sample = np.exp(1j * th_sample)
ax1.plot([0, z_sample.real], [0, z_sample.imag], 'r--o', label='Phasor at θ = π/4')

# Projections for cos(theta) and sin(theta)
ax1.plot([z_sample.real, z_sample.real], [0, z_sample.imag], color='forestgreen', linestyle=':', lw=2, label='sin(θ) [Imaginary]')
ax1.plot([0, z_sample.real], [0, 0], color='darkorange', linestyle=':', lw=2, label='cos(θ) [Real]')

ax1.set_aspect('equal')
ax1.axhline(0, color='gray', linewidth=0.8)
ax1.axvline(0, color='gray', linewidth=0.8)
ax1.set_title("Euler's Formula in the Complex Plane")
ax1.set_xlabel("Real Axis")
ax1.set_ylabel("Imaginary Axis")
ax1.legend(loc='upper right', fontsize=9)
ax1.grid(True, linestyle='--', alpha=0.6)

# Plot 2: Time-domain sinusoidal projection
t = np.linspace(0, 4 * np.pi, 500)
omega = 1.0
cosine_wave = np.cos(omega * t)
sine_wave = np.sin(omega * t)

ax2.plot(t, cosine_wave, label='Real Part: x(t) = cos(ωt)', color='darkorange', lw=2)
ax2.plot(t, sine_wave, label='Imaginary Part: sin(ωt)', color='forestgreen', linestyle='--', alpha=0.8)

ax2.set_title("Time-Domain Sinusoidal Projection")
ax2.set_xlabel("Time / Phase (rad)")
ax2.set_ylabel("Amplitude")
ax2.legend(loc='upper right', fontsize=9)
ax2.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```
