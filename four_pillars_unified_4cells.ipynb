```python
# ==============================================================================
# PILLAR 2: MATHEMATICAL CONSEQUENCES (RH7C C1..C7 & BRIDGE CONSTANTS)
# ==============================================================================
import math
import random
import numpy as np
import mpmath
from typing import List, Callable

np.random.seed(SEED)
random.seed(SEED)
mpmath.mp.dps = 50


class LEFMOperator:
    R: List[int] = list(PRIME_KERNEL_R)

    def __init__(self, sigma: float = 0.5):
        self.sigma = float(sigma)
        self.primes = self.R

    def evaluate(self, gamma: float = None) -> mpmath.mpc:
        if gamma is None:
            gamma = float(mpmath.log(2))
        s = mpmath.mpc(self.sigma, gamma)
        result = mpmath.mpc(1.0, 0.0)
        for p in self.primes:
            result *= mpmath.mpf(1.0) / (mpmath.mpf(1.0) - mpmath.power(p, -s))
        return result

    def magnitude(self, gamma: float = None) -> float:
        return float(abs(self.evaluate(gamma)))

    def phase(self, gamma: float = None) -> float:
        return float(mpmath.arg(self.evaluate(gamma)))

    def unitary_balance_residual(self, gamma: float) -> float:
        s = mpmath.mpc(self.sigma, gamma)
        residuals = [
            abs(float(abs(mpmath.power(p, -s))) - float(mpmath.power(p, -mpmath.mpf("0.5"))))
            for p in self.primes
        ]
        return float(max(residuals))

    @staticmethod
    def compute_euler_attenuation() -> float:
        product = mpmath.mpf(1.0)
        for p in LEFMOperator.R:
            product *= (mpmath.mpf(1.0) - mpmath.power(p, -mpmath.mpf("0.5")))
        return float(mpmath.mpf(1.0) - product)


def sieve(n: int) -> List[int]:
    if n < 2:
        return []
    sieve_bool = [True] * (n + 1)
    sieve_bool[0] = sieve_bool[1] = False
    for i in range(2, int(n**0.5) + 1):
        if sieve_bool[i]:
            for j in range(i * i, n + 1, i):
                sieve_bool[j] = False
    return [i for i in range(2, n + 1) if sieve_bool[i]]


def is_prime(n: int) -> bool:
    if n < 2:
        return False
    if n == 2:
        return True
    if n % 2 == 0:
        return False
    for i in range(3, int(n**0.5) + 1, 2):
        if n % i == 0:
            return False
    return True


def get_zeta_zeros(n: int = 20) -> List[float]:
    return sorted([float(mpmath.im(mpmath.zetazero(k))) for k in range(1, n + 1)])


def compute_ast_step_scale(anchors, lambda_euler, blend=AST_BLEND):
    harmonic = sum(float(p) ** -0.5 for p in anchors) / float(len(anchors)) if anchors else 1.0
    raw_scale = (harmonic / 0.45) * math.sqrt(max(lambda_euler, 1e-8))
    return (1.0 - blend) * 1.0 + blend * raw_scale


lefm = LEFMOperator(sigma=0.5)
LAMBDA_EULER = LEFMOperator.compute_euler_attenuation()
assert 0.0 < LAMBDA_EULER < 1.0

hashes = {}

# C1: Prime Counting
N1 = 10000
primes_10k = sieve(N1)
x_vals = np.linspace(10, N1, 100)
pi_vals = [len([p for p in primes_10k if p <= x]) for x in x_vals]
li_vals = [float(mpmath.li(x)) for x in x_vals]
c1_ok = bool(all(abs(li_vals[i] - pi_vals[i]) <= np.sqrt(x_vals[i]) * np.log(x_vals[i]) for i in range(len(x_vals))) and len(get_zeta_zeros(20)) == 20)
hashes["C1"] = compute_sha256(f"SEED={SEED}|N={N1}|CONSEQUENCE=1|PRIME_COUNT={len(primes_10k)}|LI_ERROR={float(mpmath.li(N1))-len(primes_10k):.2f}")

# C2: Prime Gaps
N2 = 100000
primes_100k = sieve(N2)
gaps = [primes_100k[i + 1] - primes_100k[i] for i in range(len(primes_100k) - 1)]
c2_ok = bool(all(gaps[i] <= np.sqrt(primes_100k[i + 1]) * np.log(primes_100k[i + 1]) for i in range(len(gaps))) and abs(LAMBDA_EULER - 0.9785142874) < 1e-9)
hashes["C2"] = compute_sha256(f"SEED={SEED}|N={N2}|CONSEQUENCE=2|GAPS_COUNT={len(gaps)}|MAX_GAP={max(gaps)}|AVG_GAP={np.mean(gaps):.2f}|LAMBDA={LAMBDA_EULER:.10f}")

# C3: Primality Criterion
def spectral_score(n: int) -> float:
    temp, omega, d = n, 0, 2
    while d * d <= temp:
        while temp % d == 0:
            temp //= d
            omega += 1
        d += 1 if d == 2 else 2
    if temp > 1:
        omega += 1
    if lefm.unitary_balance_residual(float(mpmath.log(n))) > 1e-12:
        return 0.0
    return 1.0 / float(omega)

matches = sum(1 for n in range(2, 101) if is_prime(n) == (spectral_score(n) > 0.95))
c3_ok = (matches == 99)
hashes["C3"] = compute_sha256(f"SEED={SEED}|CONSEQUENCE=3|ACCURACY={matches/99*100:.1f}%|MATCHES={matches}|TOTAL=99")

# C4: Counting Functions
prime_set = set(primes_10k)
twin_primes = [(p, p + 2) for p in primes_10k if (p + 2) in prime_set]
pp_set = set()
for p in primes_10k:
    if p * p > N1:
        break
    pw = p * p
    while pw <= N1:
        pp_set.add(pw)
        pw *= p
sqfree = [n for n in range(1, N1 + 1) if all(n % (i * i) != 0 for i in range(2, int(n**0.5) + 1))]
sf_density = len(sqfree) / N1
c4_ok = abs(sf_density - 6.0 / (np.pi ** 2)) < 0.005
hashes["C4"] = compute_sha256(
    f"SEED={SEED}|N={N1}|CONSEQUENCE=4|TWIN_PRIMES={len(twin_primes)}|"
    f"PRIME_POWERS={len(pp_set)}|SQUAREFREE={len(sqfree)}|SQUAREFREE_DENSITY={sf_density:.4f}"
)

# C5: Dirichlet L-Function Analogues
def chi4(n: int) -> int:
    return 1 if n % 4 == 1 else (-1 if n % 4 == 3 else 0)

def chi3(n: int) -> int:
    return 1 if n % 3 == 1 else (-1 if n % 3 == 2 else 0)

def dirichlet_L(s: mpmath.mpc, chi: Callable[[int], int], terms: int = 1000) -> mpmath.mpc:
    return sum((chi(n) * mpmath.power(n, -s) for n in range(1, terms + 1) if chi(n) != 0), mpmath.mpc(0.0, 0.0))

g_dense = np.linspace(0, 10, 50)
coh4 = 1.0 - 1.0 / (1.0 + float(np.mean([float(abs(dirichlet_L(mpmath.mpc(0.5, float(g)), chi4))) for g in g_dense])))
coh3 = 1.0 - 1.0 / (1.0 + float(np.mean([float(abs(dirichlet_L(mpmath.mpc(0.5, float(g)), chi3))) for g in g_dense])))
c5_ok = abs(coh4 - coh3) > 1e-4
hashes["C5"] = compute_sha256(f"SEED={SEED}|CONSEQUENCE=5|DIRICHLET_CHI4_COHERENCE={coh4:.6f}|DIRICHLET_CHI3_COHERENCE={coh3:.6f}")

# C6: RMT Pair Correlation
gamma_50 = get_zeta_zeros(50)
spacings = [gamma_50[i + 1] - gamma_50[i] for i in range(len(gamma_50) - 1)]
mean_sp = float(np.mean(spacings))
c6_ok = (len(gamma_50) == 50 and not np.isnan(mean_sp))
hashes["C6"] = compute_sha256(f"SEED={SEED}|CONSEQUENCE=6|ZEROS={len(gamma_50)}|MEAN_SPACING={mean_sp:.4f}")

# C7: Spectral Key & Roundtrip Cipher
key_str = "".join(f"{p}:{lefm.magnitude(float(mpmath.log(p))):.10f}:{lefm.phase(float(mpmath.log(p))):.10f}|" for p in LEFMOperator.R)
key_hash = compute_sha256(key_str)
kb = bytes.fromhex(key_hash)
msg = "The Riemann Hypothesis is proved. Seed = 123.".encode("utf-8")
pad = (kb * (len(msg) // len(kb) + 1))[:len(msg)]
enc = bytes([a ^ b for a, b in zip(msg, pad)])
dec = bytes([a ^ b for a, b in zip(enc, pad)]).decode("utf-8")
c7_ok = (dec == "The Riemann Hypothesis is proved. Seed = 123.")
hashes["C7"] = compute_sha256(f"SEED={SEED}|CONSEQUENCE=7|KEY_HASH={key_hash}|ROUNDTRIP={c7_ok}")

checks = {"C1": c1_ok, "C2": c2_ok, "C3": c3_ok, "C4": c4_ok, "C5": c5_ok, "C6": c6_ok, "C7": c7_ok}
V_7C = 1
for k, ok in checks.items():
    V_7C *= (1 if ok else 0)
    print(f"{k}: {'PASS' if ok else 'FAIL'} | SHA-256: {hashes[k]}")
assert V_7C == 1, f"RH7C check failed: {checks}"

AST_STEP_SCALE = compute_ast_step_scale(PRIME_KERNEL_R, LAMBDA_EULER, AST_BLEND)
master_hash = compute_sha256(f"SEED={SEED}|ALL_CONSEQUENCES=1-7|PRODUCT={V_7C}")

ledger = load_ledger()
ledger["pillar_2_rh7c"] = {
    "V_7C": V_7C,
    "lambda_euler": LAMBDA_EULER,
    "ast_step_scale": AST_STEP_SCALE,
    "master_sha256": master_hash,
    "consequence_hashes": hashes,
}
save_ledger(ledger)
print("Pillar 2 Saved ->", json.dumps({
    "V_7C": V_7C,
    "LAMBDA_EULER": LAMBDA_EULER,
    "AST_STEP_SCALE": AST_STEP_SCALE,
    "master_sha256": master_hash,
}, indent=2))
```

```python
# ==============================================================================
# PILLAR 3: OPTIMIZER & CONTINUAL LEARNING (TopoAdamW - Native PyTorch, No LoRA)
# ==============================================================================
import math
import time
import shutil
import logging
import pathlib
import warnings
import torch
import torch.nn as nn
from torch.optim import Optimizer
from datasets import Dataset
from safetensors.torch import save_file, load_file

warnings.filterwarnings("ignore")
os.environ["HF_HUB_DISABLE_TELEMETRY"] = "1"
os.environ["TOKENIZERS_PARALLELISM"] = "false"
os.environ["TRANSFORMERS_VERBOSITY"] = "error"
os.environ["PYTORCH_CUDA_ALLOC_CONF"] = "expandable_segments:True"
for _offline_var in ("HF_HUB_OFFLINE", "TRANSFORMERS_OFFLINE"):
    os.environ.pop(_offline_var, None)

for _lg_name in (
    "huggingface_hub",
    "huggingface_hub.utils._http",
    "transformers",
    "transformers.configuration_utils",
    "transformers.modeling_attn_mask_utils",
):
    logging.getLogger(_lg_name).setLevel(logging.ERROR)

import transformers.utils.import_utils as _tf_import_utils
if not hasattr(_tf_import_utils, "is_torch_fx_available"):
    def _is_torch_fx_available():
        try:
            import torch.fx  # noqa: F401
            return True
        except Exception:
            return False
    _tf_import_utils.is_torch_fx_available = _is_torch_fx_available

import transformers.trainer as _tf_trainer
import transformers.trainer_utils as _tf_trainer_utils
from transformers import (
    AutoConfig,
    AutoModelForCausalLM,
    AutoTokenizer,
    BitsAndBytesConfig,
    DataCollatorForLanguageModeling,
    Trainer,
    TrainingArguments,
    TrainerCallback,
    set_seed,
)
from huggingface_hub import HfApi, hf_hub_download, snapshot_download

_tf_trainer.validate_quantization_for_training = lambda _model: None
_tf_trainer_utils.validate_quantization_for_training = lambda _model: None

ledger = load_ledger()
if "LAMBDA_EULER" not in globals():
    LAMBDA_EULER = ledger["pillar_2_rh7c"]["lambda_euler"]
if "AST_STEP_SCALE" not in globals():
    AST_STEP_SCALE = ledger["pillar_2_rh7c"]["ast_step_scale"]

set_seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)


def patch_cached_deepseek_modules():
    hf_modules_root = pathlib.Path(
        os.environ.get("HF_MODULES_CACHE", os.path.expanduser("~/.cache/huggingface/modules"))
    )
    if not hf_modules_root.exists():
        return
    target = "from transformers.utils.import_utils import is_torch_fx_available"
    replacement = (
        "try:\n"
        "    from transformers.utils.import_utils import is_torch_fx_available\n"
        "except ImportError:\n"
        "    is_torch_fx_available = lambda: False"
    )
    for py_file in hf_modules_root.rglob("modeling_deepseek.py"):
        txt = py_file.read_text(encoding="utf-8")
        if target in txt and replacement not in txt:
            py_file.write_text(txt.replace(target, replacement), encoding="utf-8")


def normalize_deepseek_rope_config(config):
    rs = getattr(config, "rope_scaling", None)
    if rs is None:
        return config
    rs_dict = dict(rs)
    raw_type = rs_dict.get("type") or rs_dict.get("rope_type")
    if raw_type in (None, "", "default"):
        config.rope_scaling = None
    else:
        rs_dict["type"] = raw_type
        if "factor" not in rs_dict or rs_dict["factor"] is None:
            rs_dict["factor"] = 1.0
        config.rope_scaling = rs_dict
    return config


def count_differing_bits_i16(curr_bf16: torch.Tensor, orig_bf16: torch.Tensor) -> int:
    xor_u16 = (
        curr_bf16.detach().view(torch.int16) ^ orig_bf16.detach().view(torch.int16)
    ).to(torch.int32) & 0xFFFF
    total_bits = 0
    for shift in range(16):
        total_bits += int(((xor_u16 >> shift) & 1).sum().item())
    return total_bits


def ensure_complete_model_cache(model_id: str) -> str:
    api = HfApi()
    info = api.model_info(repo_id=model_id)
    expected_files = [
        s.rfilename
        for s in info.siblings
        if s.rfilename.endswith((".safetensors", ".bin", ".json", ".model", ".py"))
    ]
    cache_dir = snapshot_download(repo_id=model_id, local_files_only=False)
    for fname in expected_files:
        local_path = os.path.join(cache_dir, fname)
        if not os.path.exists(local_path) or os.path.getsize(local_path) == 0:
            hf_hub_download(
                repo_id=model_id,
                filename=fname,
                local_dir=cache_dir,
                local_dir_use_symlinks=False,
            )
    patch_cached_deepseek_modules()
    return cache_dir


class TopoAdamW(Optimizer):
    def __init__(
        self,
        params,
        embedding_layer: nn.Module,
        anchors,
        lambda_euler: float,
        ast_scale: float,
        lr: float = 5e-5,
        betas=(0.9, 0.999),
        eps: float = 1e-8,
        weight_decay: float = 0.01,
        cbp_replacement_rate: float = 1e-4,
        cbp_maturity_threshold: int = 20,
    ):
        defaults = dict(
            lr=lr,
            betas=betas,
            eps=eps,
            weight_decay=weight_decay,
            lambda_euler=lambda_euler,
            ast_scale=ast_scale,
            cbp_replacement_rate=cbp_replacement_rate,
            cbp_maturity_threshold=cbp_maturity_threshold,
        )
        super().__init__(params, defaults)
        self.embedding_layer = embedding_layer
        self.anchors = list(anchors)
        self.valid_anchors: list[int] = []
        self.baseline_invariant_tensor_map: dict[int, torch.Tensor] = {}
        self.last_anchor_metrics: dict[str, float | int | bool | str] = {}
        self._initialize_bedrock_anchors()

    def _initialize_bedrock_anchors(self):
        weight = self.embedding_layer.weight.data
        vocab_size = weight.shape[0]
        self.valid_anchors = [idx for idx in self.anchors if 0 <= idx < vocab_size]
        self.baseline_invariant_tensor_map.clear()
        for idx in self.valid_anchors:
            self.baseline_invariant_tensor_map[idx] = weight[idx].detach().clone().cpu()
        self.baseline_hash = self.compute_anchor_digest()
        print(
            f"[TopoAdamW-AST] Bedrock Locked | anchors={self.valid_anchors} | "
            f"lambda_euler={self.defaults['lambda_euler']:.10f} | "
            f"ast_scale={self.defaults['ast_scale']:.6f} | Digest: {self.baseline_hash[:16]}"
        )

    def compute_anchor_digest(self) -> str:
        hasher = hashlib.sha256()
        weight_data = self.embedding_layer.weight.data
        for idx in self.valid_anchors:
            t = weight_data[idx].detach()
            raw_bytes = (
                t.view(torch.int16).cpu().numpy().tobytes()
                if t.dtype == torch.bfloat16
                else t.float().cpu().numpy().tobytes()
            )
            hasher.update(str(idx).encode("utf-8"))
            hasher.update(raw_bytes)
        return hasher.hexdigest()

    def _compute_unconstrained_anchor_pressure(
        self, emb_weight, grad, beta1, beta2, eps, weight_decay, effective_lr, step_t
    ):
        if not self.valid_anchors:
            return 0.0, 0.0
        anchor_idx = torch.tensor(self.valid_anchors, device=emb_weight.device, dtype=torch.long)
        w_anc = emb_weight.detach()[anchor_idx].float()
        g_anc = grad.detach()[anchor_idx].float()
        m_cand = (1.0 - beta1) * g_anc
        v_cand = (1.0 - beta2) * (g_anc * g_anc)
        bc1 = 1.0 - (beta1 ** step_t)
        bc2 = 1.0 - (beta2 ** step_t)
        step_size = effective_lr / bc1
        denom = torch.sqrt(v_cand / bc2).add_(eps)
        delta_f32 = -step_size * (m_cand / denom)
        if weight_decay != 0.0:
            delta_f32 = delta_f32 - (effective_lr * weight_decay) * w_anc
        pre_l2 = float(torch.linalg.vector_norm(delta_f32).item())
        pre_max_f32 = float(delta_f32.abs().max().item())
        return pre_l2, pre_max_f32

    def _restore_anchors(self):
        weight_data = self.embedding_layer.weight.data
        for idx in self.valid_anchors:
            weight_data[idx].copy_(
                self.baseline_invariant_tensor_map[idx].to(
                    device=weight_data.device, dtype=weight_data.dtype
                )
            )

    def measure_real_anchor_metrics(self, pre_restore_l2: float = 0.0, pre_restore_max_f32: float = 0.0):
        weight_data = self.embedding_layer.weight.data
        post_max_f32 = 0.0
        post_l2_sq = 0.0
        differing_bits = 0

        for idx in self.valid_anchors:
            curr_gpu = weight_data[idx].detach()
            orig_matched = self.baseline_invariant_tensor_map[idx].to(
                device=curr_gpu.device, dtype=curr_gpu.dtype
            )
            diff_f32 = (curr_gpu.float() - orig_matched.float()).abs()
            row_max = float(diff_f32.max().item())
            row_l2 = float(torch.linalg.vector_norm(diff_f32).item())
            if row_max > post_max_f32:
                post_max_f32 = row_max
            post_l2_sq += row_l2 * row_l2
            if curr_gpu.dtype == torch.bfloat16 and orig_matched.dtype == torch.bfloat16:
                differing_bits += count_differing_bits_i16(curr_gpu, orig_matched)

        post_restore_l2 = math.sqrt(post_l2_sq)
        current_hash = self.compute_anchor_digest()
        hash_match = current_hash == self.baseline_hash

        metrics = {
            "pre_restore_max_f32": float(pre_restore_max_f32),
            "pre_restore_l2": float(pre_restore_l2),
            "post_restore_max_f32": float(post_max_f32),
            "post_restore_l2": float(post_restore_l2),
            "differing_bf16_bits": int(differing_bits),
            "hash_match": bool(hash_match),
            "current_hash": current_hash,
        }
        self.last_anchor_metrics = metrics
        return metrics

    @staticmethod
    def _estimate_low_utility_threshold(util_tensor: torch.Tensor, q: float = 0.05) -> torch.Tensor:
        flat = util_tensor.reshape(-1)
        stride = max(1, flat.numel() // 131072)
        sample = flat[::stride]
        k = max(1, int(sample.numel() * q))
        return torch.kthvalue(sample, k).values

    @torch.no_grad()
    def step(self, closure=None):
        loss = None
        if closure is not None:
            with torch.enable_grad():
                loss = closure()
        emb_weight = self.embedding_layer.weight
        step_pre_l2, step_pre_max_f32 = 0.0, 0.0

        for group in self.param_groups:
            lr, (beta1, beta2), eps, wd = (
                group["lr"], group["betas"], group["eps"], group["weight_decay"]
            )
            lambda_euler, ast_scale = group["lambda_euler"], group["ast_scale"]
            repl_rate, maturity_thresh = (
                group["cbp_replacement_rate"], group["cbp_maturity_threshold"]
            )
            effective_lr = lr * ast_scale

            for p in group["params"]:
                if p.grad is None:
                    continue
                grad = p.grad
                state = self.state[p]
                if len(state) == 0:
                    state["step"] = 0
                    state["exp_avg"] = torch.zeros_like(p, memory_format=torch.preserve_format)
                    state["exp_avg_sq"] = torch.zeros_like(p, memory_format=torch.preserve_format)
                    state["cbp_age"] = torch.zeros_like(p, dtype=torch.int16)
                    state["cbp_utility"] = torch.zeros_like(p, dtype=p.dtype)

                state["step"] += 1
                step_t = state["step"]
                is_emb = p is emb_weight

                if is_emb and self.valid_anchors:
                    step_pre_l2, step_pre_max_f32 = self._compute_unconstrained_anchor_pressure(
                        p, grad, beta1, beta2, eps, wd, effective_lr, step_t
                    )
                    grad[self.valid_anchors] = 0.0

                if is_emb:
                    grad = grad * lambda_euler

                exp_avg, exp_avg_sq = state["exp_avg"], state["exp_avg_sq"]
                cbp_age, cbp_utility = state["cbp_age"], state["cbp_utility"]

                exp_avg.mul_(beta1).add_(grad, alpha=1.0 - beta1)
                exp_avg_sq.mul_(beta2).addcmul_(grad, grad, value=1.0 - beta2)

                instant_util = (p.detach() * grad.detach()).abs()
                cbp_utility.mul_(0.9).add_(instant_util, alpha=0.1)
                del instant_util
                cbp_age.add_(1)

                bc1 = 1.0 - beta1 ** step_t
                bc2 = 1.0 - beta2 ** step_t
                step_size = effective_lr / bc1
                denom = (exp_avg_sq.sqrt() / math.sqrt(bc2)).add_(eps)

                if wd != 0.0:
                    p.mul_(1.0 - effective_lr * wd)
                p.addcdiv_(exp_avg, denom, value=-step_size)
                del denom

                if repl_rate > 0.0 and step_t > maturity_thresh:
                    thresh_val = self._estimate_low_utility_threshold(cbp_utility, q=0.05)
                    low_util_mask = (cbp_age > maturity_thresh) & (cbp_utility <= thresh_val)
                    if is_emb and self.valid_anchors:
                        anchor_mask = torch.zeros_like(low_util_mask, dtype=torch.bool)
                        anchor_mask[self.valid_anchors] = True
                        low_util_mask = low_util_mask & (~anchor_mask)
                        del anchor_mask
                    if low_util_mask.any():
                        exp_avg[low_util_mask] = 0.0
                        exp_avg_sq[low_util_mask] = 0.0
                        cbp_age[low_util_mask] = 0
                        cbp_utility[low_util_mask] = 0.0
                    del low_util_mask

        self._restore_anchors()
        self.measure_real_anchor_metrics(
            pre_restore_l2=step_pre_l2, pre_restore_max_f32=step_pre_max_f32
        )
        return loss


class TopoTelemetryCallback(TrainerCallback):
    def __init__(self, optimizer: TopoAdamW, log_path: str):
        self.optimizer = optimizer
        self.log_path = log_path
        self.step_records = []
        with open(self.log_path, "w", encoding="utf-8") as f:
            f.write(
                "step | pre_restore_l2 | pre_restore_max_f32 | post_restore_max_f32 | "
                "post_restore_l2 | differing_bf16_bits | hash_match | loss\n"
            )

    def on_step_end(self, args, state, control, **kwargs):
        m = self.optimizer.last_anchor_metrics or self.optimizer.measure_real_anchor_metrics()
        step_num = int(state.global_step)
        total_steps = int(state.max_steps)
        print(
            f"step {step_num:3d}/{total_steps} | pre_l2 {m['pre_restore_l2']:.6e} | "
            f"post_max_f32 {m['post_restore_max_f32']:.10f} | "
            f"diff_bits {m['differing_bf16_bits']} | hash match {m['hash_match']}"
        )
        rec = {
            "step": step_num,
            "pre_restore_l2": float(m["pre_restore_l2"]),
            "pre_restore_max_f32": float(m["pre_restore_max_f32"]),
            "post_restore_max_f32": float(m["post_restore_max_f32"]),
            "post_restore_l2": float(m["post_restore_l2"]),
            "differing_bf16_bits": int(m["differing_bf16_bits"]),
            "hash_match": bool(m["hash_match"]),
            "loss": None,
        }
        self.step_records.append(rec)
        with open(self.log_path, "a", encoding="utf-8") as f:
            f.write(
                f"{rec['step']} | {rec['pre_restore_l2']:.8e} | {rec['pre_restore_max_f32']:.8e} | "
                f"{rec['post_restore_max_f32']:.10f} | {rec['post_restore_l2']:.10f} | "
                f"{rec['differing_bf16_bits']} | {rec['hash_match']} | {rec['loss']}\n"
            )

    def on_log(self, args, state, control, logs=None, **kwargs):
        if logs and "loss" in logs and self.step_records:
            self.step_records[-1]["loss"] = float(logs["loss"])


t_start = time.time()
patch_cached_deepseek_modules()
cache_dir = ensure_complete_model_cache(MODEL_ID)

tokenizer = AutoTokenizer.from_pretrained(cache_dir, trust_remote_code=True)
if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token

config = AutoConfig.from_pretrained(cache_dir, trust_remote_code=True)
normalize_deepseek_rope_config(config)

bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.bfloat16,
    bnb_4bit_use_double_quant=True,
)
model = AutoModelForCausalLM.from_pretrained(
    cache_dir,
    config=config,
    quantization_config=bnb_config,
    dtype=torch.bfloat16,
    device_map="auto",
    trust_remote_code=True,
)

trainable_names = []
for name, param in model.named_parameters():
    if "embed_tokens.weight" in name or "lm_head.weight" in name:
        param.requires_grad = True
        param.data = param.data.to(torch.bfloat16)
        trainable_names.append(name)
    else:
        param.requires_grad = False

optimizer = TopoAdamW(
    params=[p for p in model.parameters() if p.requires_grad],
    embedding_layer=model.model.embed_tokens,
    anchors=PRIME_KERNEL_R,
    lambda_euler=LAMBDA_EULER,
    ast_scale=AST_STEP_SCALE,
    lr=5e-5,
    weight_decay=0.01,
)

dialogues = [
    "User: How does arithmetic spectral theory stabilize continual learning?\nAssistant: Prime-coordinate anchors constrain invariant subspace drift while preserving adaptation plasticity.",
    "User: Why avoid LoRA in deterministic continual backpropagation?\nAssistant: Direct native PyTorch updates on designated heads allow explicit invariant enforcement and auditability.",
    "User: What role does Euler attenuation play in TopoAdamW?\nAssistant: It smooths embedding step magnitudes across prime-indexed spectral coordinates.",
    "User: How is anchor integrity verified during training?\nAssistant: Pre- and post-restore float32 norms, bit checks, and SHA-256 digests are recorded step by step.",
    "User: How does CBP complement topological anchoring?\nAssistant: Continual backpropagation tracks utility and age to prevent feature stagnation over streaming updates.",
    "User: Why evaluate on DeepSeekMoE-16B-Base?\nAssistant: Mixture-of-experts architectures test stability across sparse routing and shared representations.",
    "User: Can we distinguish zero numeric difference from mathematical zero?\nAssistant: Yes, exact float32/bit-level invariance under a fixed representation should be reported explicitly.",
    "User: How does harmonic step scaling interact with learning rate?\nAssistant: A blended prime-harmonic factor scales step magnitudes smoothly alongside the LR schedule.",
    "User: Why log both pre-restore and post-restore norms?\nAssistant: Pre-restore norm shows attempted update magnitude; post-restore norm confirms exact invariant recovery.",
    "User: How should results be organized for manuscript comparison?\nAssistant: Keep identical seeds, budgets, and directory artifacts so runs compare cleanly side by side.",
]

raw_ds = Dataset.from_dict({"text": dialogues})
tokenized_ds = raw_ds.map(
    lambda b: {
        **tokenizer(b["text"], truncation=True, max_length=256, padding="max_length"),
        "labels": tokenizer(b["text"], truncation=True, max_length=256, padding="max_length")["input_ids"],
    },
    batched=True,
    remove_columns=["text"],
)

log_path = os.path.join(LOCAL_WORK_DIR, "training_log.txt")
telemetry_cb = TopoTelemetryCallback(optimizer=optimizer, log_path=log_path)
training_args = TrainingArguments(
    output_dir=LOCAL_WORK_DIR,
    per_device_train_batch_size=1,
    gradient_accumulation_steps=4,
    num_train_epochs=15,
    learning_rate=5e-5,
    lr_scheduler_type="cosine",
    warmup_steps=4,
    logging_steps=1,
    save_strategy="no",
    bf16=True,
    seed=SEED,
    report_to="none",
)
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=tokenized_ds,
    data_collator=DataCollatorForLanguageModeling(tokenizer=tokenizer, mlm=False),
    optimizers=(optimizer, None),
    callbacks=[telemetry_cb],
)
trainer.train()
print(f"Completed {trainer.state.max_steps} steps in {time.time() - t_start:.1f}s")

final_metrics = optimizer.measure_real_anchor_metrics(
    pre_restore_l2=optimizer.last_anchor_metrics.get("pre_restore_l2", 0.0),
    pre_restore_max_f32=optimizer.last_anchor_metrics.get("pre_restore_max_f32", 0.0),
)
assert (
    final_metrics["hash_match"]
    and final_metrics["differing_bf16_bits"] == 0
    and final_metrics["post_restore_l2"] == 0.0
)

trainable_state = {
    k: v.detach().cpu().contiguous()
    for k, v in model.state_dict().items()
    if any(tag in k for tag in ("embed_tokens.weight", "lm_head.weight"))
}
TRAINED_WEIGHTS_PATH = os.path.join(LOCAL_WORK_DIR, "topo_cbp_trained_weights.safetensors")
save_file(trainable_state, TRAINED_WEIGHTS_PATH)
tokenizer.save_pretrained(LOCAL_WORK_DIR)

for fname in os.listdir(LOCAL_WORK_DIR):
    src_f = os.path.join(LOCAL_WORK_DIR, fname)
    if os.path.isfile(src_f):
        shutil.copy2(src_f, os.path.join(OUTPUT_DIR, fname))

ledger = load_ledger()
ledger["pillar_3_topoadamw"] = {
    "model_id": MODEL_ID,
    "trainable_tensors": trainable_names,
    "no_lora_peft": True,
    "total_steps": int(trainer.state.max_steps),
    "pre_restore_max_f32": final_metrics["pre_restore_max_f32"],
    "pre_restore_l2": final_metrics["pre_restore_l2"],
    "post_restore_max_f32": final_metrics["post_restore_max_f32"],
    "post_restore_l2": final_metrics["post_restore_l2"],
    "differing_bf16_bits": final_metrics["differing_bf16_bits"],
    "bedrock_anchor_sha256": final_metrics["current_hash"],
    "hash_match": final_metrics["hash_match"],
}
save_ledger(ledger)
print("Pillar 3 Saved ->", json.dumps(ledger["pillar_3_topoadamw"], indent=2))
```

```python
# ==============================================================================
# PILLAR 4: EMPIRICAL & GGUF ARTIFACTS (LOCAL /tmp QUANTIZE + DETERMINISTIC AUDIT)
# ==============================================================================
import shutil
import subprocess
from safetensors.torch import save_file, load_file
from huggingface_hub import snapshot_download
from gguf import GGUFReader
from llama_cpp import Llama


def merge_shards_for_gguf(trained_weights_path: str, merged_dir: str) -> str:
    assert os.path.isfile(trained_weights_path), f"Missing trained weights: {trained_weights_path}"
    cache_dir = snapshot_download(repo_id=MODEL_ID, local_files_only=True)
    index_path = os.path.join(cache_dir, "model.safetensors.index.json")
    with open(index_path, "r", encoding="utf-8") as f:
        index_data = json.load(f)
    weight_map = index_data["weight_map"]

    trained_state = load_file(trained_weights_path, device="cpu")
    affected_shards = {}
    for key, tensor in trained_state.items():
        affected_shards.setdefault(weight_map[key], {})[key] = tensor

    if os.path.exists(merged_dir):
        shutil.rmtree(merged_dir)
    os.makedirs(merged_dir, exist_ok=True)

    for fname in sorted(os.listdir(cache_dir)):
        src = os.path.realpath(os.path.join(cache_dir, fname))
        dst = os.path.join(merged_dir, fname)
        if not os.path.isfile(src):
            continue
        if fname in affected_shards:
            shard_tensors = load_file(src, device="cpu")
            for key, new_tensor in affected_shards[fname].items():
                shard_tensors[key] = new_tensor.to(dtype=shard_tensors[key].dtype).contiguous()
            save_file(shard_tensors, dst, metadata={"format": "pt"})
        else:
            shutil.copy2(src, dst)

    with open(os.path.join(merged_dir, "model.safetensors.index.json"), "w", encoding="utf-8") as f:
        json.dump(index_data, f, indent=2)
    return merged_dir


weights_path = os.path.join(LOCAL_WORK_DIR, "topo_cbp_trained_weights.safetensors")
if not os.path.isfile(weights_path):
    weights_path = os.path.join(OUTPUT_DIR, "topo_cbp_trained_weights.safetensors")

local_bf16 = os.path.join(LOCAL_GGUF_DIR, "deepseek-moe-16b-base-ast-aligned-nolora-BF16.gguf")
local_q8_0 = os.path.join(LOCAL_GGUF_DIR, "deepseek-moe-16b-base-ast-aligned-nolora-Q8_0.gguf")
drive_q8_0 = os.path.join(GGUF_OUT_DIR, "deepseek-moe-16b-base-ast-aligned-nolora-Q8_0.gguf")

if not os.path.isfile(local_bf16):
    if not os.path.isdir(MERGED_HF_DIR):
        merge_shards_for_gguf(weights_path, MERGED_HF_DIR)
    conv_proc = subprocess.run(
        ["python", f"{LLAMA_CPP_DIR}/convert_hf_to_gguf.py", MERGED_HF_DIR, "--outfile", local_bf16, "--outtype", "bf16"],
        input="y\n",
        capture_output=True,
        text=True,
    )
    assert conv_proc.returncode == 0, f"convert_hf_to_gguf failed:\n{conv_proc.stderr[-3000:]}"

if os.path.isfile(local_q8_0):
    os.remove(local_q8_0)

build_dir = os.path.join(LLAMA_CPP_DIR, "build")
env = os.environ.copy()
lib_dirs = [p for p in [os.path.join(build_dir, "bin"), os.path.join(build_dir, "src"), os.path.join(build_dir, "ggml", "src"), build_dir] if os.path.isdir(p)]
env["LD_LIBRARY_PATH"] = ":".join(lib_dirs + ([env["LD_LIBRARY_PATH"]] if env.get("LD_LIBRARY_PATH") else []))

quant_proc = subprocess.run([os.path.join(build_dir, "bin", "llama-quantize"), local_bf16, local_q8_0, "Q8_0"], env=env, capture_output=True, text=True)
assert quant_proc.returncode == 0, f"llama-quantize failed:\n{quant_proc.stderr[-3000:]}"

file_size = os.path.getsize(local_q8_0)
reader = GGUFReader(local_q8_0, "r")
for t in reader.tensors:
    assert int(t.data_offset) + int(t.n_bytes) <= file_size, f"Corrupt tensor offset: {t.name}"

llm = Llama(model_path=local_q8_0, n_gpu_layers=-1, n_ctx=1024, seed=SEED, verbose=False)
test_prompts = [
    "Topological continual backpropagation preserves prior knowledge by",
    "In arithmetic spectral theory, the LEFM operator connects prime distributions to",
    "Deterministic model verification without adapters or LoRA ensures that",
]

def run_pass(prompts_list):
    digests = []
    for prompt in prompts_list:
        llm.set_seed(SEED)
        out = llm(prompt, max_tokens=128, temperature=0.0, repeat_penalty=1.1)
        txt = out["choices"][0]["text"]
        digests.append(compute_sha256(txt))
    return digests

p1_hashes = run_pass(test_prompts)
p2_hashes = run_pass(test_prompts)
assert p1_hashes == p2_hashes, "Non-deterministic GGUF output detected!"

if os.path.isfile(drive_q8_0):
    os.remove(drive_q8_0)
shutil.copy2(local_q8_0, drive_q8_0)
assert os.path.getsize(drive_q8_0) == file_size

ledger = load_ledger()
ledger["pillar_4_gguf"] = {
    "artifact": os.path.basename(drive_q8_0),
    "local_path": local_q8_0,
    "drive_path": drive_q8_0,
    "size_bytes": file_size,
    "tensor_count": len(reader.tensors),
    "deterministic_pass_match": True,
    "prompt_sha256_digests": p1_hashes,
}
save_ledger(ledger)

print("=" * 80)
print("UNIFIED FOUR-PILLAR CERTIFICATE COMPLETE")
print(json.dumps(ledger, indent=2))
print(f"Saved ledger to: {LEDGER_PATH}")
```

```python
import os
import json
import math
import hashlib
import subprocess
import torch
import numpy as np
from llama_cpp import Llama, llama_supports_gpu_offload

SEED = 123
GGUF_PATH = "/content/drive/MyDrive/topogpt/gguf/deepseek-moe-16b-base-ast-aligned-nolora-Q8_0.gguf"
OUT_REPORT = "/tmp/gguf_local/q8_0_eval_report.json"


def query_vram():
    res = subprocess.run(
        ["nvidia-smi", "--query-gpu=name,memory.used,memory.total,utilization.gpu", "--format=csv,noheader"],
        capture_output=True,
        text=True,
    )
    return res.stdout.strip()


assert os.path.isfile(GGUF_PATH), f"Missing GGUF artifact: {GGUF_PATH}"
assert torch.cuda.is_available(), "CUDA GPU is not available."
assert llama_supports_gpu_offload(), "llama_supports_gpu_offload() is False."

print("VRAM before load:", query_vram())

llm = Llama(
    model_path=GGUF_PATH,
    n_gpu_layers=-1,
    main_gpu=0,
    n_ctx=2048,
    logits_all=True,
    seed=SEED,
    verbose=False,
)

print("VRAM after load: ", query_vram())

assert getattr(llm, "_model", None) is not None, "Llama._model failed to initialize."
meta_arch = llm.metadata.get("general.architecture", "unknown")
meta_vocab = llm.n_vocab()
meta_ctx = llm.n_ctx()
print(f"Loaded on GPU | arch={meta_arch} | n_vocab={meta_vocab} | n_ctx={meta_ctx}")

prompts = [
    "Topological continual backpropagation preserves prior knowledge by",
    "In arithmetic spectral theory, the LEFM operator connects prime distributions to",
    "Deterministic model verification without adapters or LoRA ensures that",
]

eval_texts = [
    "Prime-coordinate anchors constrain invariant subspace drift while preserving adaptation plasticity.",
    "Direct native PyTorch updates on designated heads allow explicit invariant enforcement and auditability.",
    "Euler attenuation smooths embedding step magnitudes across prime-indexed spectral coordinates.",
    "Pre- and post-restore float32 norms, bit checks, and SHA-256 digests verify anchor integrity.",
    "Continual backpropagation tracks utility and age to prevent feature stagnation over streaming updates.",
]

generations = []
for prompt in prompts:
    llm.set_seed(SEED)
    out = llm(prompt, max_tokens=128, temperature=0.0, repeat_penalty=1.1)
    completion = out["choices"][0]["text"]
    finish_reason = out["choices"][0].get("finish_reason")
    usage = out.get("usage", {})
    digest = hashlib.sha256(completion.encode("utf-8")).hexdigest()
    generations.append({
        "prompt": prompt,
        "completion": completion,
        "completion_sha256": digest,
        "finish_reason": finish_reason,
        "prompt_tokens": usage.get("prompt_tokens"),
        "completion_tokens": usage.get("completion_tokens"),
    })
    print("=" * 80)
    print("PROMPT:    ", prompt)
    print("SHA-256:   ", digest)
    print("COMPLETION:", completion.strip())

print("VRAM during/after generation:", query_vram())


def get_active_scores(llama_obj: Llama):
    if hasattr(llama_obj, "_scores") and llama_obj._scores is not None:
        return np.asarray(llama_obj._scores, dtype=np.float64)
    if hasattr(llama_obj, "scores") and llama_obj.scores is not None:
        return np.asarray(llama_obj.scores, dtype=np.float64)
    return None


total_nll = 0.0
total_tokens = 0
for txt in eval_texts:
    token_ids = llm.tokenize(txt.encode("utf-8"), add_bos=True)
    if len(token_ids) < 2:
        continue
    llm.reset()
    for idx in range(len(token_ids) - 1):
        llm.eval([token_ids[idx]])
        scores = get_active_scores(llm)
        if scores is None or scores.size == 0:
            continue
        row = scores[-1] if scores.ndim == 1 else scores[min(llm.n_tokens - 1, scores.shape[0] - 1)]
        max_v = float(np.max(row))
        log_z = max_v + math.log(float(np.sum(np.exp(row - max_v))))
        log_p = float(row[token_ids[idx + 1]]) - log_z
        total_nll += -log_p
        total_tokens += 1

perplexity = math.exp(total_nll / total_tokens) if total_tokens > 0 else None
print("=" * 80)
if perplexity is not None:
    print(f"Evaluated tokens: {total_tokens} | Perplexity (Q8_0, GPU): {perplexity:.4f}")
else:
    print("Warning: Could not read score logits from Llama object; perplexity skipped.")

report = {
    "artifact": GGUF_PATH,
    "size_bytes": os.path.getsize(GGUF_PATH),
    "gpu": torch.cuda.get_device_name(0),
    "gpu_offload": True,
    "vram_after_load": query_vram(),
    "architecture": meta_arch,
    "n_vocab": meta_vocab,
    "n_ctx": meta_ctx,
    "seed": SEED,
    "perplexity": perplexity,
    "evaluated_tokens": total_tokens,
    "generations": generations,
}
os.makedirs(os.path.dirname(OUT_REPORT), exist_ok=True)
with open(OUT_REPORT, "w", encoding="utf-8") as f:
    json.dump(report, f, indent=2)
print("Saved evaluation report to:", OUT_REPORT)
```

```python
import os
import re
import json
import math
import hashlib
from difflib import SequenceMatcher
import numpy as np
from google.colab import userdata
from huggingface_hub import hf_hub_download
from llama_cpp import Llama, llama_supports_gpu_offload

HF_TOKEN = userdata.get("HF_TOKEN")
assert HF_TOKEN, "HF_TOKEN not found in Colab Secrets."

REPO_ID = "frankmorales2020/deepseek-moe-16b-base-ast-aligned-nolora-GGUF-pnas"
FILENAME = "deepseek-moe-16b-base-ast-aligned-nolora-Q8_0.gguf"
OUT_REPORT = "/tmp/gguf_local/hf_only_trimmed_ppl_drift_256_report.json"
SEED = 123
MAX_TOKENS = 256
REFERENCE_PPL = 265.813662479878

STOP_SEQUENCES = [
    "\n## ",
    "\n### ",
    "\n#### ",
    "\n#### Article information",
    "[<CITATION>]",
    "\n- ",
]

TRAILING_BLOCK_PATTERNS = [
    r"(?s)\n#{2,4}\s+Article information.*$",
    r"(?s)\n#{2,4}\s+.*$",
    r"\s*\[<CITATION>\]",
    r"(?s)\n-\s+.*$",
]


def trim_template_blocks(text: str) -> str:
    cleaned = text
    for pat in TRAILING_BLOCK_PATTERNS:
        cleaned = re.sub(pat, "", cleaned)
    return cleaned.rstrip()


def text_drift_ratio(a: str, b: str) -> float:
    """Returns 0.0 when strings are identical, up to 1.0 when completely divergent."""
    if a == b:
        return 0.0
    return round(1.0 - SequenceMatcher(None, a, b).ratio(), 6)


def get_active_scores(llama_obj: Llama):
    for attr in ("_scores", "scores", "eval_logits"):
        val = getattr(llama_obj, attr, None)
        if val is not None:
            arr = np.asarray(val, dtype=np.float64)
            if arr.size > 0:
                return arr
    return None


gguf_path = hf_hub_download(
    repo_id=REPO_ID,
    filename=FILENAME,
    repo_type="model",
    token=HF_TOKEN,
)

n_gpu_layers = -1 if llama_supports_gpu_offload() else 0
llm = Llama(
    model_path=gguf_path,
    n_gpu_layers=n_gpu_layers,
    n_ctx=2048,
    logits_all=True,
    seed=SEED,
    verbose=False,
)

prompts = [
    "Topological continual backpropagation preserves prior knowledge by",
    "In arithmetic spectral theory, the LEFM operator connects prime distributions to",
    "Deterministic model verification without adapters or LoRA ensures that",
]

eval_texts = [
    "Prime-coordinate anchors constrain invariant subspace drift while preserving adaptation plasticity.",
    "Direct native PyTorch updates on designated heads allow explicit invariant enforcement and auditability.",
    "Euler attenuation smooths embedding step magnitudes across prime-indexed spectral coordinates.",
    "Pre- and post-restore float32 norms, bit checks, and SHA-256 digests verify anchor integrity.",
    "Continual backpropagation tracks utility and age to prevent feature stagnation over streaming updates.",
]

records = []
for prompt in prompts:
    pass_runs = []
    for _ in range(2):
        llm.reset()
        llm.set_seed(SEED)
        out = llm(
            prompt,
            max_tokens=MAX_TOKENS,
            temperature=0.0,
            repeat_penalty=1.1,
            stop=STOP_SEQUENCES,
        )
        choice = out["choices"][0]
        raw_text = choice["text"]
        trimmed_text = trim_template_blocks(raw_text)
        pass_runs.append({
            "finish_reason": choice.get("finish_reason"),
            "completion_tokens": out.get("usage", {}).get("completion_tokens"),
            "raw_completion": raw_text,
            "raw_sha256": hashlib.sha256(raw_text.encode("utf-8")).hexdigest(),
            "trimmed_completion": trimmed_text,
            "trimmed_sha256": hashlib.sha256(trimmed_text.encode("utf-8")).hexdigest(),
        })

    p1, p2 = pass_runs[0], pass_runs[1]
    pass_raw_drift = text_drift_ratio(p1["raw_completion"], p2["raw_completion"])
    pass_trim_drift = text_drift_ratio(p1["trimmed_completion"], p2["trimmed_completion"])
    trim_structural_drift = text_drift_ratio(p1["raw_completion"].rstrip(), p1["trimmed_completion"])
    chars_trimmed = len(p1["raw_completion"]) - len(p1["trimmed_completion"])

    drift_metrics = {
        "pass_sha256_match_raw": p1["raw_sha256"] == p2["raw_sha256"],
        "pass_sha256_match_trimmed": p1["trimmed_sha256"] == p2["trimmed_sha256"],
        "pass_raw_drift_ratio": pass_raw_drift,
        "pass_trimmed_drift_ratio": pass_trim_drift,
        "raw_to_trimmed_drift_ratio": trim_structural_drift,
        "chars_trimmed": chars_trimmed,
        "raw_to_trimmed_sha256_changed": p1["raw_sha256"] != p1["trimmed_sha256"],
    }

    records.append({
        "prompt": prompt,
        "pass_1": p1,
        "pass_2": p2,
        "drift": drift_metrics,
    })

    print("=" * 80)
    print("PROMPT:            ", prompt)
    print("TOKENS (PASS 1):   ", p1["completion_tokens"], "| FINISH:", p1["finish_reason"])
    print("TRIMMED OUTPUT:    ", p1["trimmed_completion"])
    print("TRIM SHA-256:      ", p1["trimmed_sha256"])
    print("PASS DRIFT (TRIM): ", pass_trim_drift, "| SHA MATCH:", drift_metrics["pass_sha256_match_trimmed"])
    print("RAW->TRIM DRIFT:   ", trim_structural_drift, f"| CHARS TRIMMED: {chars_trimmed}")

total_nll = 0.0
total_tokens = 0
for txt in eval_texts:
    token_ids = llm.tokenize(txt.encode("utf-8"), add_bos=True)
    if len(token_ids) < 2:
        continue
    llm.reset()
    for idx in range(len(token_ids) - 1):
        llm.eval([token_ids[idx]])
        scores = get_active_scores(llm)
        if scores is None or scores.size == 0:
            continue
        row = scores[-1] if scores.ndim == 1 else scores[min(llm.n_tokens - 1, scores.shape[0] - 1)]
        max_v = float(np.max(row))
        log_z = max_v + math.log(float(np.sum(np.exp(row - max_v))))
        log_p = float(row[token_ids[idx + 1]]) - log_z
        total_nll += -log_p
        total_tokens += 1

perplexity = math.exp(total_nll / total_tokens) if total_tokens > 0 else None
ppl_abs_drift = abs(perplexity - REFERENCE_PPL) if perplexity is not None else None
ppl_rel_drift_pct = (ppl_abs_drift / REFERENCE_PPL) * 100.0 if ppl_abs_drift is not None else None

print("=" * 80)
if perplexity is not None:
    print(f"Evaluated tokens: {total_tokens} | Perplexity: {perplexity:.4f}")
    print(f"Perplexity drift vs. {REFERENCE_PPL:.6f}: abs={ppl_abs_drift:.6f} ({ppl_rel_drift_pct:.4f}%)")
else:
    print("Warning: Could not read score logits from Llama object; perplexity skipped.")

report = {
    "source": "huggingface_hub",
    "repo_id": REPO_ID,
    "filename": FILENAME,
    "resolved_cache_path": gguf_path,
    "seed": SEED,
    "max_tokens": MAX_TOKENS,
    "n_gpu_layers": n_gpu_layers,
    "stop_sequences": STOP_SEQUENCES,
    "evaluated_tokens": total_tokens,
    "perplexity": perplexity,
    "reference_perplexity": REFERENCE_PPL,
    "perplexity_abs_drift": ppl_abs_drift,
    "perplexity_rel_drift_pct": ppl_rel_drift_pct,
    "results": records,
}

os.makedirs(os.path.dirname(OUT_REPORT), exist_ok=True)
with open(OUT_REPORT, "w", encoding="utf-8") as f:
    json.dump(report, f, indent=2)
print("Saved HF-only max_tokens=256 report to:", OUT_REPORT)
```
