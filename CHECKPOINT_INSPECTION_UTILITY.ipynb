```python
#!/usr/bin/env python3
# ============================================================================
# AUDIT INSPECTION UTILITY FOR THE REMAINING 4 FLEET CHECKPOINTS
# REPOS:
#  1. frankmorales2020/topo-cbp-world-model-3d
#  2. frankmorales2020/qwen2.5-0.5b-topo-governed-cbp-wikitext
#  3. frankmorales2020/topo-cbp-rlhf-2026
#  4. frankmorales2020/muse-glimmer-30b-topo-governed-cbp
# ============================================================================

import os
import re
import math
import torch
from huggingface_hub import list_repo_files, hf_hub_download
from safetensors import safe_open

try:
    from google.colab import userdata
    HF_TOKEN = userdata.get("HF_TOKEN")
except Exception:
    HF_TOKEN = os.environ.get("HF_TOKEN", None)

TARGET_REPOS = [
    "frankmorales2020/topo-cbp-world-model-3d",
    "frankmorales2020/qwen2.5-0.5b-topo-governed-cbp-wikitext",
    "frankmorales2020/topo-cbp-rlhf-2026",
    "frankmorales2020/muse-glimmer-30b-topo-governed-cbp",
]


def audit_repository(repo_id: str):
    print("=" * 86)
    print(f"🏛️  SOVEREIGN MACHINE LABORATORY (SOMALA) — BARE-METAL AUDIT")
    print(f"📦  REPOSITORY: {repo_id}")
    print("=" * 86)

    try:
        files = list_repo_files(repo_id=repo_id, token=HF_TOKEN)
    except Exception as e:
        print(f"❌ Failed to access repository manifest: {e}\n")
        return

    print("📁 Repository Manifest:")
    for f in files:
        print(f"  • {f}")
    print("-" * 86)

    # 1. Locate primary weight container
    weight_candidates = [
        "model.safetensors",
        "pytorch_model.bin",
        "topo_cbp_model.pt",
        "checkpoint.pt",
    ]
    target_weight = None
    for cand in weight_candidates:
        if cand in files:
            target_weight = cand
            break

    if not target_weight:
        weights = [f for f in files if f.endswith((".safetensors", ".pt", ".bin"))]
        target_weight = weights[0] if weights else None

    # 2. Inspect Weights & Tensors
    if target_weight:
        print(f"📥 Auditing Weight Container: {target_weight}")
        try:
            ckpt_path = hf_hub_download(repo_id=repo_id, filename=target_weight, token=HF_TOKEN)

            if target_weight.endswith(".safetensors"):
                with safe_open(ckpt_path, framework="pt", device="cpu") as sf:
                    keys = list(sf.keys())
                    total_tensors = len(keys)
                    total_params = sum(math.prod(sf.get_slice(k).get_shape()) for k in keys)
                    metadata = sf.metadata() or {}
                print(f"  • Container Format : safetensors")
                print(f"  • Tensor Count     : {total_tensors} tensors")
                print(f"  • Parameter Count  : {total_params / 1e6:.2f}M ({total_params / 1e9:.2f}B)")
                if metadata:
                    print(f"  • Header Metadata  : {metadata}")

            elif target_weight.endswith((".pt", ".bin")):
                state = torch.load(ckpt_path, map_location="cpu", weights_only=False)
                if isinstance(state, dict):
                    print(f"  • Container Format : PyTorch State / Metadata Dict")
                    print(f"  • Root Keys Count  : {len(state)}")

                    # Check for direct tensor tensors vs nested state
                    tensors = {k: v for k, v in state.items() if isinstance(v, torch.Tensor)}
                    if tensors:
                        total_params = sum(v.numel() for v in tensors.values())
                        print(f"  • Tensor Count     : {len(tensors)} tensors")
                        print(f"  • Parameter Count  : {total_params / 1e6:.2f}M")

                    # Check for direct or nested topo_invariants
                    topo = state.get("topo_invariants") or state.get("topo_metadata")
                    if isinstance(topo, dict):
                        print("\n  🔒 Embedded Topological Invariants:")
                        for tk, tv in topo.items():
                            print(f"     • {tk:24s} : {tv}")
                    else:
                        for field in ["prime_ring", "prime_anchors", "lambda_euler", "safety_constant", "delta_drift", "sha256_digest", "state_hash"]:
                            if field in state:
                                print(f"     • {field:24s} : {state[field]}")
                else:
                    print(f"  • Object Type      : {type(state)}")

        except Exception as e:
            print(f"⚠️ Error while deserializing weight container: {e}")
    else:
        print("⚠️ No direct binary weights (.safetensors/.pt/.bin) identified.")

    # 3. Audit README.md for Topological Governance Receipts
    if "README.md" in files:
        print("-" * 86)
        print("📋 Model Card Invariant Receipts (README.md):")
        try:
            readme_path = hf_hub_download(repo_id=repo_id, filename="README.md", token=HF_TOKEN)
            with open(readme_path, "r", encoding="utf-8") as rf:
                readme_text = rf.read()

            doi = re.search(r"(10\.5281/zenodo\.[0-9]+)", readme_text)
            if doi:
                print(f"  • Zenodo Verification DOI : {doi.group(1)}")

            primes = re.search(r"(\\mathcal\{P\}|P)\s*=\s*\{([0-9,\s]+)\}", readme_text)
            if primes:
                print(f"  • Prime Ring Anchors       : {{{primes.group(2)}}}")

            euler = re.search(r"\\Lambda\s*=\s*[^=]+=\s*([0-9\.]+)", readme_text)
            if euler:
                print(f"  • Euler Constant (Λ)       : {euler.group(1)}")

            # Check for drift metrics
            drift_matches = re.findall(r"(\b0\.0000000000\b|drift\s*[:=]\s*[0-9\.]+)", readme_text, re.IGNORECASE)
            if drift_matches:
                print(f"  • Drift Metric Statement   : {drift_matches[0]}")

            # Extract any summary table rows
            table_rows = re.findall(r"\|\s*\*\*?([0-9]+)\*\*?\s*\|\s*([0-9\.]+)\s*\|\s*\*\*?`?([0-9\.]+)`?\*\*?.*\|", readme_text)
            if table_rows:
                print(f"\n  📊 Logged Streaming Updates Sample (Total Logged: {len(table_rows)}):")
                print(f"     {'Step':<8} | {'Loss':<10} | {'Drift'}")
                print(f"     " + "-" * 36)
                for r in table_rows[:3]:
                    print(f"     {r[0]:<8} | {r[1]:<10} | {r[2]}")
                if len(table_rows) > 3:
                    last = table_rows[-1]
                    print(f"     ... ({len(table_rows)-4} steps omitted)")
                    print(f"     {last[0]:<8} | {last[1]:<10} | {last[2]}")

        except Exception as e:
            print(f"⚠️ Error parsing README.md: {e}")

    print("=" * 86 + "\n")


if __name__ == "__main__":
    for repo in TARGET_REPOS:
        audit_repository(repo)
```

```python
#!/usr/bin/env python3
# ============================================================================
# SOVEREIGN MACHINE LABORATORY (SOMALA) — FLEET INSPECTION UTILITY
# GITHUB REPO : frank-morales2020/AST
# SCRIPT      : CHECKPOINT_INSPECTION_UTILITY.py
#
# Audits the Topo-CBP fleet for prime coordinate invariance (P = {2,3,5,7,11,13}),
# Euler attenuation (Λ ≈ 0.9785), zero parameter drift (Δ_drift = 0.0),
# multi-task continuum heads, and hardware-level cryptographic state digests.
# ============================================================================

import os
import re
import math
import json
import torch
from huggingface_hub import list_repo_files, hf_hub_download
from safetensors import safe_open

# 1. Environment & HF Hub Authentication
try:
    from google.colab import userdata
    HF_TOKEN = userdata.get("HF_TOKEN")
except Exception:
    HF_TOKEN = os.environ.get("HF_TOKEN", None)

FLEET_SPECS = [
    {
        "repo_id": "frankmorales2020/qwen2.5-0.5b-topo-governed-cbp-fineweb",
        "description": "1D Causal LM Streaming Horizon (FineWeb-Edu)",
        "target_weight": "model.safetensors",
    },
    {
        "repo_id": "frankmorales2020/topo-cbp-world-model-3d",
        "description": "3D Latent Spatial Dynamics & World Model Engine",
        "target_weight": "model.safetensors",
    },
    {
        "repo_id": "frankmorales2020/qwen2.5-0.5b-topo-governed-cbp-wikitext",
        "description": "1D Language Benchmark with Bedrock State Anchor",
        "target_weight": "model.safetensors",
    },
    {
        "repo_id": "frankmorales2020/topo-cbp-rlhf-2026",
        "description": "21B Aligned Continual Reinforcement Substrate (RLHF/PPO)",
        "target_weight": "topo_cbp_best.pt",
    },
    {
        "repo_id": "frankmorales2020/topo-cbp-gemma-4-e4b-cifar100-certified",
        "description": "Gemma-4 Continual Vision Benchmark (CIFAR-100 Certified)",
        "target_weight": "topo_cbp_gemma4_cifar100_best_certified.pt",
    },
    {
        "repo_id": "frankmorales2020/topo-gemma-4-e4b-vision-15tasks-video",
        "description": "15-Task Multimodal Continuum (2D Vision, 3D CT, 4D Video)",
        "target_weight": "pytorch_model.bin",
    },
    {
        "repo_id": "frankmorales2020/muse-glimmer-30b-topo-governed-cbp",
        "description": "30B Frontier Parameter Certified Topological Checkpoint",
        "target_weight": "certified_topological_best.pt",
    },
]


def audit_checkpoint(spec: dict):
    repo_id = spec["repo_id"]
    desc = spec["description"]
    target_filename = spec["target_weight"]

    print("=" * 90)
    print(f"🏛️  SOVEREIGN MACHINE LABORATORY (SOMALA) — BARE-METAL AUDIT")
    print(f"📦  REPOSITORY  : {repo_id}")
    print(f"🎯  MODALITY    : {desc}")
    print("=" * 90)

    # 1. Fetch File Manifest
    try:
        files = list_repo_files(repo_id=repo_id, token=HF_TOKEN)
        print(f"📁 Manifest: {len(files)} files found")
        for f in files:
            print(f"   • {f}")
    except Exception as e:
        print(f"❌ Failed to access repository manifest: {e}")
        return

    print("-" * 90)

    # 2. Checkpoint Resolution & Download
    if target_filename not in files:
        # Fallback to any valid binary
        candidates = [f for f in files if f.endswith((".safetensors", ".pt", ".bin"))]
        target_filename = candidates[0] if candidates else None

    if not target_filename:
        print("⚠️ No serialized model weights found in repository.")
        return

    print(f"📥 Auditing Weight Container: {target_filename}")
    try:
        ckpt_path = hf_hub_download(repo_id=repo_id, filename=target_filename, token=HF_TOKEN)
    except Exception as e:
        print(f"❌ Download failed: {e}")
        return

    # 3. Unpack Weights & Invariants by Container Type
    if target_filename.endswith(".safetensors"):
        try:
            with safe_open(ckpt_path, framework="pt", device="cpu") as sf:
                tensor_keys = list(sf.keys())
                total_tensors = len(tensor_keys)
                total_params = sum(math.prod(sf.get_slice(k).get_shape()) for k in tensor_keys)
                st_meta = sf.metadata() or {}

            print(f"  • Container Standard       : safetensors ({st_meta.get('format', 'pt')})")
            print(f"  • Total Serialized Tensors : {total_tensors} tensors")
            print(f"  • Parameter Count          : {total_params / 1e6:.2f}M ({total_params / 1e9:.2f}B)")
            if len(st_meta) > 1:
                print(f"  • Header Metadata          : {st_meta}")
        except Exception as e:
            print(f"⚠️ Failed to parse safetensors file: {e}")

    elif target_filename.endswith((".pt", ".bin")):
        try:
            state = torch.load(ckpt_path, map_location="cpu", weights_only=False)
            if isinstance(state, dict):
                print(f"  • Container Standard       : PyTorch State / Metadata Dictionary")
                print(f"  • Root Keys Count          : {len(state)}")

                # Check parameter count if raw tensors present
                tensors = {k: v for k, v in state.items() if isinstance(v, torch.Tensor)}
                if tensors:
                    total_params = sum(v.numel() for v in tensors.values())
                    print(f"  • Parameter Count          : {total_params / 1e6:.2f}M ({total_params / 1e9:.2f}B)")
                    print(f"  • Direct Weight Tensors    : {len(tensors)} tensors")

                # Nested topo_invariants dictionary inspection
                topo = state.get("topo_invariants") or state.get("topo_metadata")
                if isinstance(topo, dict):
                    print("\n  🔒 Embedded Topological Invariants [topo_invariants]:")
                    for k, v in topo.items():
                        print(f"     • {k:24s} : {v}")

                # Root-level governance keys
                gov_keys = [
                    "prime_ring", "prime_anchors", "lambda_euler", "safety_constant",
                    "delta_drift", "drift", "sha256_digest", "state_hash", "forgetting_avg"
                ]
                root_gov = {k: state[k] for k in gov_keys if k in state}
                if root_gov:
                    print("\n  🔒 Root Governance Key Signatures:")
                    for k, v in root_gov.items():
                        print(f"     • {k:24s} : {v}")

                # Multi-task heads audit (for 15-task multimodal continuum)
                if "classifiers" in state and isinstance(state["classifiers"], dict):
                    print(f"\n  🎯 Multimodal Classifiers   : {len(state['classifiers'])} heads resident")
                    for tid, tweights in list(state["classifiers"].items())[:5]:
                        if isinstance(tweights, dict):
                            summary = [f"{k}: {list(v.shape)}" for k, v in tweights.items() if isinstance(v, torch.Tensor)]
                            print(f"     Task {tid:<4s} -> Sub-layers: {', '.join(summary)}")
                    if len(state["classifiers"]) > 5:
                        print(f"     ... ({len(state['classifiers']) - 5} additional task heads verified)")

                if "boundary_layer" in state:
                    print(f"  ⚡ Latent Boundary Layer   : Layer {state.get('boundary_layer')} (d_model = {state.get('hidden_size', 'N/A')})")

            else:
                print(f"  • State Object Type        : {type(state)}")
        except Exception as e:
            print(f"⚠️ Failed to parse PyTorch state dictionary: {e}")

    # 4. Audit Model Card (README.md) Receipts & Tables
    if "README.md" in files:
        try:
            readme_path = hf_hub_download(repo_id=repo_id, filename="README.md", token=HF_TOKEN)
            with open(readme_path, "r", encoding="utf-8") as rf:
                text = rf.read()

            doi = re.search(r"(10\.5281/zenodo\.[0-9]+)", text)
            primes = re.search(r"(\\mathcal\{P\}|P)\s*=\s*\{([0-9,\s]+)\}", text)
            euler = re.search(r"\\Lambda\s*=\s*[^=]+=\s*([0-9\.]+)", text)
            drift = re.search(r"(0\.0000000000|drift\s*[:=]\s*[0-9\.]+)", text, re.IGNORECASE)

            print("\n📋 Serialized Model Card Receipts (README.md):")
            if doi:
                print(f"  • Zenodo Verification DOI  : {doi.group(1)}")
            if primes:
                print(f"  • Prime Coordinate Ring    : {{{primes.group(2)}}}")
            if euler:
                print(f"  • Euler Attenuation (Λ)    : {euler.group(1)}")
            if drift:
                print(f"  • Certified Drift Metric   : {drift.group(1)}")

            # Check for embedded comparative drift tables
            table_rows = re.findall(
                r"\|\s*\*\*?(\d+)\*\*?\s*\|\s*([0-9\.]+)\s*\|\s*\*\*?`?([0-9\.]+)`?\*\*?\s*\|\s*([0-9\.]+)\s*\|\s*\*\*?`?([0-9\.]+)`?\*\*?\s*\|\s*([^|]+)\|",
                text
            )
            if table_rows:
                print(f"\n📊 Streaming Horizon Audit Sample ({len(table_rows)} Total Logged Intervals):")
                print(f"   {'Step':<7} | {'Gov Loss':<10} | {'Gov Drift':<16} | {'Base Loss':<10} | {'Base Drift':<12} | {'Phenomenon'}")
                print("   " + "-" * 76)
                for r in table_rows[:3]:
                    print(f"   {r[0]:<7} | {r[1]:<10} | {r[2]:<16} | {r[3]:<10} | {r[4]:<12} | {r[5].strip()}")
                if len(table_rows) > 3:
                    last = table_rows[-1]
                    print(f"   ... ({len(table_rows) - 4} intermediate updates verified)")
                    print(f"   {last[0]:<7} | {last[1]:<10} | {last[2]:<16} | {last[3]:<10} | {last[4]:<12} | {last[5].strip()}")

        except Exception as e:
            print(f"⚠️ Error parsing README.md receipts: {e}")

    print("=" * 90 + "\n")


if __name__ == "__main__":
    for spec in FLEET_SPECS:
        audit_checkpoint(spec)
```
